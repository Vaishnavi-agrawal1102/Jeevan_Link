```python

# STEP 6: PREPROCESS THE INDOFLOODS PRECIPITATION DATASET

import pandas as pd
from pathlib import Path

# 1. Define the input and output paths.
raw_path = Path(
    "../../Raw_Data/INDOFLOOD_Data/precipitation_variables_indofloods.csv"
)
output_path = Path("precipitation_variables_indofloods_processed.csv")

# 2. Load the original dataset.
df_precip = pd.read_csv(raw_path, low_memory=False)

# 3. Create a working copy to protect the raw data.
df_precip_processed = df_precip.copy()

# 4. Standardize column names by trimming extra spaces.
df_precip_processed.columns = (
    df_precip_processed.columns.astype(str).str.strip()
)

# 5. Trim extra spaces from EventID.
df_precip_processed["EventID"] = (
    df_precip_processed["EventID"].astype("string").str.strip()
)

# 6. Convert precipitation measurements to numeric values.
precipitation_columns = [
    f"T{i}d" for i in range(1, 11)
]

for column in precipitation_columns:
    df_precip_processed[column] = pd.to_numeric(
        df_precip_processed[column],
        errors="coerce"
    )

# 7. Identify and remove completely empty columns.
empty_columns = df_precip_processed.columns[
    df_precip_processed.isna().all()
].tolist()

df_precip_processed = df_precip_processed.drop(
    columns=empty_columns
)

# 8. Remove exact duplicate rows.
duplicates_removed = int(
    df_precip_processed.duplicated().sum()
)

df_precip_processed = (
    df_precip_processed.drop_duplicates().copy()
)

# 9. Display the processing results.
print("Original dataset shape:", df_precip.shape)
print("Completely empty columns removed:", empty_columns)
print("Exact duplicate rows removed:", duplicates_removed)
print("Processed dataset shape:", df_precip_processed.shape)

print("\nMissing values after processing:")
print(df_precip_processed.isna().sum())

print("\nData types after processing:")
print(df_precip_processed.dtypes)

# 10. Save the processed dataset.
df_precip_processed.to_csv(output_path, index=False)

print("\nProcessed dataset saved to:")
print(output_path.resolve())

# 11. Save a preprocessing report.
report_path = Path(
    "precipitation_variables_preprocessing_report.txt"
)

with open(report_path, "w", encoding="utf-8") as report:
    report.write("INDOFLOODS PRECIPITATION PREPROCESSING REPORT\n")
    report.write(f"Original shape: {df_precip.shape}\n")
    report.write(f"Empty columns removed: {empty_columns}\n")
    report.write(f"Duplicate rows removed: {duplicates_removed}\n")
    report.write(
        f"Processed shape: {df_precip_processed.shape}\n"
    )
    report.write("\nMissing values after processing:\n")
    report.write(
        df_precip_processed.isna().sum().to_string()
    )

print("Preprocessing report saved to:")
print(report_path.resolve())
```

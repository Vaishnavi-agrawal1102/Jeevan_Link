```python

# STEP 4: PREPROCESS INDOFLOODS CATCHMENT CHARACTERISTICS

import pandas as pd
from pathlib import Path

# 1. Define the raw input and processed output paths.
raw_path = Path(
    "../../Raw_Data/INDOFLOOD_Data/catchment_characteristics_indofloods.csv"
)
output_path = Path("catchment_characteristics_indofloods_processed.csv")

# 2. Load the original dataset.
df_catchment = pd.read_csv(raw_path, low_memory=False)

# 3. Create a working copy to protect the original file.
df_catchment_processed = df_catchment.copy()

# 4. Standardize column names by removing leading and trailing spaces.
df_catchment_processed.columns = (
    df_catchment_processed.columns.astype(str).str.strip()
)

# 5. Trim extra spaces in text columns without changing missing values.
for column in df_catchment_processed.select_dtypes(
    include=["object", "string"]
).columns:
    df_catchment_processed[column] = (
        df_catchment_processed[column].map(
            lambda value: value.strip()
            if isinstance(value, str)
            else value
        )
    )

# 6. Identify columns that are completely empty.
# Keep their names in the report before removing them.
empty_columns = df_catchment_processed.columns[
    df_catchment_processed.isna().all()
].tolist()

# Remove only columns that contain no values anywhere.
df_catchment_processed = df_catchment_processed.drop(
    columns=empty_columns
)

# 7. Remove exact duplicate rows.
duplicates_removed = int(
    df_catchment_processed.duplicated().sum()
)

df_catchment_processed = (
    df_catchment_processed.drop_duplicates().copy()
)

# 8. Display the results.
print("Original dataset shape:", df_catchment.shape)
print("Completely empty columns removed:", empty_columns)
print("Exact duplicate rows removed:", duplicates_removed)
print("Processed dataset shape:", df_catchment_processed.shape)

print("\nMissing values after processing:")
print(
    df_catchment_processed.isna()
    .sum()
    .sort_values(ascending=False)
    .head(20)
)

print("\nData types:")
print(df_catchment_processed.dtypes)

# 9. Save the processed dataset.
df_catchment_processed.to_csv(output_path, index=False)

print("\nProcessed dataset saved to:")
print(output_path.resolve())

# 10. Save a preprocessing report.
report_path = Path(
    "catchment_characteristics_preprocessing_report.txt"
)

with open(report_path, "w", encoding="utf-8") as report:
    report.write("INDOFLOODS CATCHMENT CHARACTERISTICS REPORT\n")
    report.write(f"Original shape: {df_catchment.shape}\n")
    report.write(f"Empty columns removed: {empty_columns}\n")
    report.write(f"Duplicate rows removed: {duplicates_removed}\n")
    report.write(
        f"Processed shape: {df_catchment_processed.shape}\n"
    )
    report.write("\nMissing values by column:\n")
    report.write(
        df_catchment_processed.isna()
        .sum()
        .sort_values(ascending=False)
        .to_string()
    )

print("Preprocessing report saved to:")
print(report_path.resolve())
```

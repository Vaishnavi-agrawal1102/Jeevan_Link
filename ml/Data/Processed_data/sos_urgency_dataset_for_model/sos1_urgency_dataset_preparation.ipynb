```python

# Import pandas for working with CSV datasets.
import pandas as pd

# Import Path to manage file and folder paths.
from pathlib import Path

# Set the location of your existing processed datasets.
base_folder = Path(".")

# Set the folder containing your preprocessed IFI datasets.
ifi_folder = base_folder / "ifi_processed_Data"

# Set the folder containing your preprocessed INDOFLOODS datasets.
indoflood_folder = base_folder / "indoflood_processed_Data"

# Set the folder containing your preprocessed Disaster Response datasets.
disaster_folder = base_folder / "Disaster Response Messages"

# List example files to inspect; update a path if your folder name differs.
files_to_check = {
    "IFI district impact": ifi_folder / "District_FloodImpact_processed.csv",
    "INDOFLOODS events": indoflood_folder / "floodevents_indofloods_processed.csv",
    "Disaster Response training": disaster_folder / "disaster_response_training_processed.csv"
}

# Print the columns available in each processed dataset.
for dataset_name, file_path in files_to_check.items():
    print(f"\n--- {dataset_name} ---")

    # Check whether the expected file exists.
    if file_path.exists():
        # Load the processed CSV.
        existing_df = pd.read_csv(file_path, low_memory=False)

        # Display the dataset dimensions and available columns.
        print("Shape:", existing_df.shape)
        print("Columns:", existing_df.columns.tolist())
    else:
        # Report the file if the path needs correcting.
        print("File not found:", file_path.resolve())

# Define the fields required for individual SOS urgency examples.
sos_columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "water_level",
    "building_condition",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
    "location_risk",
    "urgency_label"
]

# Create an empty template with the required columns.
sos_df = pd.DataFrame(columns=sos_columns)

# Define the output file for the new SOS dataset template.
output_file = base_folder / "sos_urgency_dataset_processed.csv"

# Save the template without adding any invented training examples.
sos_df.to_csv(output_file, index=False)

# Display confirmation and the template location.
print("\nSOS urgency dataset template created.")
print("Rows:", len(sos_df))
print("Columns:", len(sos_df.columns))
print("Saved to:", output_file.resolve())

# Display the empty template.
display(sos_df)
```

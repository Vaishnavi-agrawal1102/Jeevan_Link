```python

# Import pandas to create and save the SOS data-entry file.
import pandas as pd

# Import Path to manage file locations.
from pathlib import Path

# Define the output folder.
output_folder = Path(".")

# Define the columns needed for each SOS example.
columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "water_level",
    "building_condition",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
    "location_risk",
    "urgency_label"
]

# Create an empty dataset for collecting reviewed SOS examples.
sos_df = pd.DataFrame(columns=columns)

# Set the output CSV filename.
output_file = output_folder / "sos_urgency_data_collection.csv"

# Save the empty data-entry CSV.
sos_df.to_csv(output_file, index=False)

# Confirm that the file has been created.
print("SOS data-entry file created.")
print("Columns:", len(columns))
print("Saved to:", output_file.resolve())

# Display the empty dataset structure.
display(sos_df)
```

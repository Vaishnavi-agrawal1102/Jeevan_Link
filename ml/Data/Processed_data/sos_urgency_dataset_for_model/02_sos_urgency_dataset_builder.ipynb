```python

# Step 2: Prepare to expand the synthetic SOS dataset
# This cell checks the existing data and defines the urgency rules.
# It does not overwrite or modify the original collection CSV.

import pandas as pd
from pathlib import Path

# Load the current collection
collection_file = Path("sos_urgency_data_collection.csv")
sos_df = pd.read_csv(collection_file, low_memory=False)

# Define a transparent example rubric for synthetic scenarios.
# These rules are for a project demonstration, not real emergency triage.

def assign_example_urgency(row):
    # Critical: immediate life-threatening indicators
    if (
        row["medical_emergency"] == "Yes"
        or (
            row["people_trapped"] == "Yes"
            and row["water_level"] == "Rapidly_Rising"
        )
        or (
            row["building_condition"] == "Collapsing"
            and row["people_trapped"] == "Yes"
        )
    ):
        return "Critical"

    # High: rescue needs, injuries, or substantial danger
    if (
        row["people_trapped"] == "Yes"
        or row["injuries_reported"] == "Yes"
        or row["rescue_required"] == "Yes"
        or row["location_risk"] == "High"
    ):
        return "High"

    # Medium: non-immediate needs or moderate risk
    if (
        row["location_risk"] == "Medium"
        or row["food_required"] == "Yes"
        or row["water_required"] == "Yes"
        or row["shelter_required"] == "Yes"
        or row["building_condition"] == "Damaged"
    ):
        return "Medium"

    # Low: no immediate danger indicators in this example rubric
    return "Low"


# Show the current records and their rubric-based labels side by side.
# This lets us inspect disagreements before generating more data.
sos_df["rubric_example_label"] = sos_df.apply(
    assign_example_urgency, axis=1
)

print("Existing labels compared with the example rubric:")
display(
    sos_df[
        ["sos_id", "urgency_label", "rubric_example_label", "record_source"]
    ]
)

print("\nNumber of records:", len(sos_df))
print("\nOriginal collection file remains unchanged.")
```

```python

# STEP 3: Build a large SOS urgency dataset using the available sources
# This creates illustrative synthetic SOS scenarios for pipeline development.
# It does NOT create real or expert-verified SOS urgency labels.
# It does NOT overwrite the original collection CSV.

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# 1. Set up folders and input files
# ------------------------------------------------------------

BASE_DIR = Path.cwd()

# This notebook should be inside Data/Processed_data/sos_urgency_dataset.
# If your notebook is elsewhere, update DATA_DIR below.
DATA_DIR = BASE_DIR.parent

SOS_FILE = BASE_DIR / "sos_urgency_data_collection.csv"

IFI_DIR = DATA_DIR / "ifi_processed_Data"
INDO_DIR = DATA_DIR / "indoflood_processed_Data"
DR_DIR = DATA_DIR / "Disaster Response Messages"

OUTPUT_FILE = BASE_DIR / "sos_urgency_dataset_processed.csv"

# Read processed sources. Fail clearly if a required file is missing.
required_files = {
    "SOS collection": SOS_FILE,
    "IFI district impact": IFI_DIR / "District_FloodImpact_processed.csv",
    "IFI flooded area": IFI_DIR / "District_FloodedArea_processed.csv",
    "IFI district severity": IFI_DIR / "DFSI_processed.csv",
    "INDOFLOODS events": INDO_DIR / "floodevents_indofloods_processed.csv",
    "INDOFLOODS precipitation": INDO_DIR / "precipitation_variables_indofloods_processed.csv",
    "Disaster Response training": DR_DIR / "disaster_response_training_processed.csv",
}

missing_files = [
    f"{name}: {path}"
    for name, path in required_files.items()
    if not path.exists()
]

if missing_files:
    raise FileNotFoundError(
        "These files were not found. Check your folder names and paths:\n"
        + "\n".join(missing_files)
    )

# Load the files with low_memory=False to avoid mixed-type warnings.
sos_original = pd.read_csv(SOS_FILE, low_memory=False)
ifi_impact = pd.read_csv(required_files["IFI district impact"], low_memory=False)
ifi_area = pd.read_csv(required_files["IFI flooded area"], low_memory=False)
ifi_severity = pd.read_csv(required_files["IFI district severity"], low_memory=False)
flood_events = pd.read_csv(required_files["INDOFLOODS events"], low_memory=False)
precipitation = pd.read_csv(required_files["INDOFLOODS precipitation"], low_memory=False)
disaster_messages = pd.read_csv(required_files["Disaster Response training"], low_memory=False)

print("Source files loaded:")
print("Existing SOS records:", len(sos_original))
print("IFI district-impact rows:", len(ifi_impact))
print("IFI flooded-area rows:", len(ifi_area))
print("IFI district-severity rows:", len(ifi_severity))
print("INDOFLOODS event rows:", len(flood_events))
print("INDOFLOODS precipitation rows:", len(precipitation))
print("Disaster Response message rows:", len(disaster_messages))


# ------------------------------------------------------------
# 2. Join only datasets with a valid shared key
# ------------------------------------------------------------

# EventID is the shared key between INDOFLOODS events and precipitation.
if "EventID" not in flood_events.columns or "EventID" not in precipitation.columns:
    raise ValueError("EventID is missing from an INDOFLOODS input file.")

# Remove repeated EventIDs from the precipitation side before joining.
# This prevents accidental multiplication of event rows.
precipitation_unique = precipitation.drop_duplicates(
    subset=["EventID"], keep="first"
)

flood_context = flood_events.merge(
    precipitation_unique,
    on="EventID",
    how="left",
    suffixes=("_event", "_rain"),
    validate="one_to_one"
)

print("\nINDOFLOODS event/precipitation rows after valid EventID join:",
      len(flood_context))


# ------------------------------------------------------------
# 3. Combine IFI district tables by district name
# ------------------------------------------------------------

# These are district-level sources, not individual SOS records.
# We combine the IFI tables only on their district-name fields.
if "Dist_Name" in ifi_impact.columns and "Dist_Name" in ifi_area.columns:
    ifi_district_context = ifi_impact.merge(
        ifi_area,
        on="Dist_Name",
        how="left",
        validate="one_to_one"
    )
else:
    raise ValueError("Expected Dist_Name column was not found in IFI files.")

if (
    "District_Name" in ifi_severity.columns
    and "State_Name" in ifi_severity.columns
):
    ifi_district_context = ifi_district_context.merge(
        ifi_severity,
        left_on="Dist_Name",
        right_on="District_Name",
        how="left",
        validate="many_to_one"
    )

print("IFI district-context rows:", len(ifi_district_context))


# ------------------------------------------------------------
# 4. Helper functions for safely reading numeric source values
# ------------------------------------------------------------

rng = np.random.default_rng(42)

def numeric_value(row, columns, default=0.0):
    """Return the first available numeric value from the requested columns."""
    for column in columns:
        if column in row.index:
            value = pd.to_numeric(row[column], errors="coerce")
            if pd.notna(value):
                return float(value)
    return default

def sample_row(df):
    """Select one source row without changing the original dataframe."""
    if df.empty:
        raise ValueError("Cannot sample from an empty source dataframe.")
    return df.iloc[int(rng.integers(0, len(df)))]

def yes_no(condition):
    return "Yes" if condition else "No"


# Identify Disaster Response assistance columns that exist in this dataset.
assistance_columns = [
    "medical_help",
    "search_and_rescue",
    "water",
    "food",
    "shelter",
    "clothing",
    "transport",
    "buildings",
    "floods",
]

assistance_columns = [
    c for c in assistance_columns if c in disaster_messages.columns
]

if "message" not in disaster_messages.columns:
    raise ValueError("The Disaster Response dataset has no 'message' column.")


# ------------------------------------------------------------
# 5. Define an explicit illustrative urgency-labeling rubric
# ------------------------------------------------------------

def assign_synthetic_label(row):
    """
    Example-only rules for synthetic scenarios.
    These rules are not a validated emergency triage standard.
    """
    if (
        row["medical_emergency"] == "Yes"
        or (
            row["people_trapped"] == "Yes"
            and row["water_level"] == "Rapidly_Rising"
        )
        or (
            row["building_condition"] == "Collapsing"
            and row["people_trapped"] == "Yes"
        )
    ):
        return "Critical"

    if (
        row["people_trapped"] == "Yes"
        or row["injuries_reported"] == "Yes"
        or row["rescue_required"] == "Yes"
        or row["location_risk"] == "High"
    ):
        return "High"

    if (
        row["location_risk"] == "Medium"
        or row["food_required"] == "Yes"
        or row["water_required"] == "Yes"
        or row["shelter_required"] == "Yes"
        or row["building_condition"] == "Damaged"
    ):
        return "Medium"

    return "Low"


# ------------------------------------------------------------
# 6. Generate 5,000 synthetic SOS scenarios with source context
# ------------------------------------------------------------

N_SYNTHETIC = 5000

water_levels = ["Stable", "Rising", "Rapidly_Rising"]
building_conditions = ["Intact", "Damaged", "Collapsing", "Unknown"]
location_risks = ["Low", "Medium", "High"]

synthetic_records = []

for i in range(N_SYNTHETIC):

    # Sample contextual rows independently.
    # These samples are NOT claimed to describe the same real location/event.
    event_row = sample_row(flood_context)
    district_row = sample_row(ifi_district_context)
    message_row = sample_row(disaster_messages)

    # Use flood measurements as context, not as direct SOS urgency labels.
    peak_level = numeric_value(
        event_row, ["Peak Flood Level (m)"], default=np.nan
    )
    duration = numeric_value(
        event_row, ["Event Duration (days)"], default=np.nan
    )
    rainfall_1d = numeric_value(
        event_row, ["T1d"], default=np.nan
    )

    fatalities = numeric_value(
        district_row, ["Human_fatality"], default=np.nan
    )
    injuries = numeric_value(
        district_row, ["Human_injured"], default=np.nan
    )
    population = numeric_value(
        district_row, ["Population"], default=np.nan
    )
    flooded_percent = numeric_value(
        district_row,
        ["Corrected_Percent_Flooded_Area", "Percent_Flooded_Area"],
        default=np.nan
    )
    dfsi = numeric_value(district_row, ["DFSI"], default=np.nan)

    # Generate SOS-level details for an illustrative scenario.
    people_affected = int(rng.integers(1, 31))
    people_trapped = rng.choice(
        ["Yes", "No"], p=[0.25, 0.75]
    )
    injuries_reported = rng.choice(
        ["Yes", "No"], p=[0.30, 0.70]
    )
    medical_emergency = rng.choice(
        ["Yes", "No"], p=[0.18, 0.82]
    )

    water_level = rng.choice(
        water_levels, p=[0.30, 0.45, 0.25]
    )
    building_condition = rng.choice(
        building_conditions, p=[0.55, 0.25, 0.05, 0.15]
    )
    location_risk = rng.choice(
        location_risks, p=[0.25, 0.40, 0.35]
    )

    # Include assistance-category context from a sampled Disaster Response
    # message. These are source-message labels, not urgency labels.
    assistance_values = {}
    for column in assistance_columns:
        raw_value = pd.to_numeric(
            pd.Series([message_row[column]]), errors="coerce"
        ).iloc[0]
        assistance_values[column] = (
            int(raw_value) if pd.notna(raw_value) else np.nan
        )

    rescue_required = yes_no(
        people_trapped == "Yes"
        or injuries_reported == "Yes"
        or (
            "search_and_rescue" in assistance_values
            and assistance_values["search_and_rescue"] == 1
        )
    )

    medical_need = yes_no(
        medical_emergency == "Yes"
        or injuries_reported == "Yes"
        or (
            "medical_help" in assistance_values
            and assistance_values["medical_help"] == 1
        )
    )

    food_required = yes_no(
        assistance_values.get("food", 0) == 1
        if pd.notna(assistance_values.get("food", 0))
        else False
    )
    water_required = yes_no(
        assistance_values.get("water", 0) == 1
        if pd.notna(assistance_values.get("water", 0))
        else False
    )
    shelter_required = yes_no(
        assistance_values.get("shelter", 0) == 1
        if pd.notna(assistance_values.get("shelter", 0))
        else False
    )

    record = {
        "sos_id": f"SYN{i + 1:06d}",
        "sos_description": (
            f"Synthetic flood scenario involving {people_affected} people; "
            f"water is {water_level.lower().replace('_', ' ')}; "
            f"building condition is {building_condition.lower()}."
        ),
        "people_affected": people_affected,
        "people_trapped": people_trapped,
        "injuries_reported": injuries_reported,
        "medical_emergency": medical_need,
        "water_level": water_level,
        "building_condition": building_condition,
        "rescue_required": rescue_required,
        "food_required": food_required,
        "water_required": water_required,
        "shelter_required": shelter_required,
        "location_risk": location_risk,

        # Context from the flood-event source.
        "indoflood_event_id_context": event_row.get("EventID", np.nan),
        "context_peak_flood_level_m": peak_level,
        "context_event_duration_days": duration,
        "context_rainfall_t1d": rainfall_1d,

        # Context from the IFI district source.
        "ifi_district_context": district_row.get(
            "Dist_Name", np.nan
        ),
        "context_district_fatalities": fatalities,
        "context_district_injuries": injuries,
        "context_district_population": population,
        "context_flooded_area_percent": flooded_percent,
        "context_dfsi": dfsi,

        # Source message is retained for traceability, but it is not
        # presented as the generated SOS description or an urgency label.
        "disaster_response_message_id_context": message_row.get(
            "id", np.nan
        ),
        "disaster_response_message_context": str(
            message_row["message"]
        ),
        **{
            f"source_{column}": value
            for column, value in assistance_values.items()
        },

        # Important provenance fields.
        "record_source": "synthetic",
        "label_source": "synthetic_rule",
        "context_link_status": "independent_sample_not_linked",
    }

    record["urgency_label"] = assign_synthetic_label(record)
    synthetic_records.append(record)


synthetic_df = pd.DataFrame(synthetic_records)


# ------------------------------------------------------------
# 7. Prepare original SOS records for combining
# ------------------------------------------------------------

# Keep all original columns and rows.
# Add provenance columns if they are absent.
if "record_source" not in sos_original.columns:
    sos_original["record_source"] = "unknown"

if "label_source" not in sos_original.columns:
    sos_original["label_source"] = "existing_label_unverified"

if "context_link_status" not in sos_original.columns:
    sos_original["context_link_status"] = "not_recorded"

# Add context columns to original records as missing because we have
# not verified any event/district/message match for these SOS reports.
context_columns = [
    "indoflood_event_id_context",
    "context_peak_flood_level_m",
    "context_event_duration_days",
    "context_rainfall_t1d",
    "ifi_district_context",
    "context_district_fatalities",
    "context_district_injuries",
    "context_district_population",
    "context_flooded_area_percent",
    "context_dfsi",
    "disaster_response_message_id_context",
    "disaster_response_message_context",
]

for column in context_columns:
    if column not in sos_original.columns:
        sos_original[column] = np.nan

for column in synthetic_df.columns:
    if column not in sos_original.columns:
        sos_original[column] = np.nan

for column in sos_original.columns:
    if column not in synthetic_df.columns:
        synthetic_df[column] = np.nan

# Match column order before combining.
synthetic_df = synthetic_df[sos_original.columns]

# Combine original SOS records with newly generated synthetic scenarios.
final_df = pd.concat(
    [sos_original, synthetic_df],
    ignore_index=True
)

# Remove duplicate SOS IDs if any exist; preserve the original record first.
final_df = final_df.drop_duplicates(
    subset=["sos_id"], keep="first"
).reset_index(drop=True)


# ------------------------------------------------------------
# 8. Validate and save the prepared dataset
# ------------------------------------------------------------

valid_labels = ["Low", "Medium", "High", "Critical"]

print("\nFINAL DATASET SUMMARY")
print("---------------------")
print("Rows:", len(final_df))
print("Columns:", len(final_df.columns))
print("Duplicate SOS IDs:", final_df["sos_id"].duplicated().sum())
print("Missing SOS IDs:", final_df["sos_id"].isna().sum())
print("Invalid urgency labels:", (~final_df["urgency_label"].isin(valid_labels)).sum())

print("\nRecords by source:")
display(final_df["record_source"].value_counts(dropna=False))

print("\nUrgency label distribution:")
display(
    final_df["urgency_label"]
    .value_counts(dropna=False)
    .reindex(valid_labels, fill_value=0)
)

print("\nMissing values in the most important columns:")
important_columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "urgency_label",
    "record_source",
    "label_source",
]
display(final_df[important_columns].isna().sum().to_frame("Missing_Count"))

# Save to a new file; do not overwrite the collection file.
final_df.to_csv(OUTPUT_FILE, index=False)

print("\nSaved processed dataset to:")
print(OUTPUT_FILE.resolve())
print("\nOriginal collection CSV was not overwritten.")
```

```python

# ============================================================
# JEEVAN LINK: SOS URGENCY DATASET BUILDER
# ============================================================
# This notebook prepares a large dataset for the later training
# stage. Generated scenarios and their labels are synthetic.
#
# IMPORTANT:
# - Source flood data is contextual, not verified SOS data.
# - IFI district tables are kept separate to avoid unsafe joins.
# - INDOFLOODS events and precipitation are joined by EventID.
# - This is a project-development dataset, not a validated
#   real-world emergency triage dataset.
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# 1. Configure paths
# ------------------------------------------------------------

# Expected location:
# Data/Processed_data/sos_urgency_dataset/
#     02_sos_urgency_dataset_builder.ipynb

BASE_DIR = Path.cwd()
DATA_DIR = BASE_DIR.parent

SOS_FILE = BASE_DIR / "sos_urgency_data_collection.csv"

IFI_DIR = DATA_DIR / "ifi_processed_Data"
INDO_DIR = DATA_DIR / "indoflood_processed_Data"
DR_DIR = DATA_DIR / "Disaster Response Messages"

OUTPUT_FILE = BASE_DIR / "sos_urgency_dataset_processed.csv"

required_files = {
    "SOS collection": SOS_FILE,
    "IFI impact": IFI_DIR / "District_FloodImpact_processed.csv",
    "IFI flooded area": IFI_DIR / "District_FloodedArea_processed.csv",
    "IFI severity": IFI_DIR / "DFSI_processed.csv",
    "INDOFLOODS events": INDO_DIR / "floodevents_indofloods_processed.csv",
    "INDOFLOODS precipitation": INDO_DIR / "precipitation_variables_indofloods_processed.csv",
    "Disaster Response": DR_DIR / "disaster_response_training_processed.csv",
}

missing_files = [
    f"{name}: {path}"
    for name, path in required_files.items()
    if not path.exists()
]

if missing_files:
    raise FileNotFoundError(
        "Some input files were not found.\n"
        "Check your folder structure and filenames:\n"
        + "\n".join(missing_files)
    )

print("All required input files were found.")


# ------------------------------------------------------------
# 2. Load the processed source datasets
# ------------------------------------------------------------

sos_original = pd.read_csv(SOS_FILE, low_memory=False)

ifi_impact = pd.read_csv(
    required_files["IFI impact"], low_memory=False
)

ifi_area = pd.read_csv(
    required_files["IFI flooded area"], low_memory=False
)

ifi_severity = pd.read_csv(
    required_files["IFI severity"], low_memory=False
)

flood_events = pd.read_csv(
    required_files["INDOFLOODS events"], low_memory=False
)

precipitation = pd.read_csv(
    required_files["INDOFLOODS precipitation"], low_memory=False
)

disaster_messages = pd.read_csv(
    required_files["Disaster Response"], low_memory=False
)

print("\nSOURCE DATASET SIZES")
print("Existing SOS records:", len(sos_original))
print("IFI impact records:", len(ifi_impact))
print("IFI flooded-area records:", len(ifi_area))
print("IFI severity records:", len(ifi_severity))
print("INDOFLOODS event records:", len(flood_events))
print("INDOFLOODS precipitation records:", len(precipitation))
print("Disaster Response messages:", len(disaster_messages))


# ------------------------------------------------------------
# 3. Validate required columns
# ------------------------------------------------------------

required_sos_columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "water_level",
    "building_condition",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
    "location_risk",
    "urgency_label",
]

missing_sos_columns = [
    c for c in required_sos_columns
    if c not in sos_original.columns
]

if missing_sos_columns:
    raise ValueError(
        f"SOS collection is missing columns: {missing_sos_columns}"
    )

if "EventID" not in flood_events.columns:
    raise ValueError("EventID is missing from the INDOFLOODS event file.")

if "EventID" not in precipitation.columns:
    raise ValueError("EventID is missing from the precipitation file.")

if "message" not in disaster_messages.columns:
    raise ValueError("The Disaster Response file has no message column.")


# ------------------------------------------------------------
# 4. Join INDOFLOODS using EventID
# ------------------------------------------------------------

# Remove duplicate precipitation EventIDs before joining.
# This avoids multiplying event rows during the merge.
precipitation_unique = precipitation.drop_duplicates(
    subset=["EventID"],
    keep="first"
)

flood_context = flood_events.merge(
    precipitation_unique,
    on="EventID",
    how="left",
    suffixes=("_event", "_rain"),
    validate="one_to_one"
)

print("\nINDOFLOODS event/context rows:", len(flood_context))


# ------------------------------------------------------------
# 5. Keep IFI tables separate
# ------------------------------------------------------------

# District names repeat in IFI, so merging on Dist_Name alone
# could combine unrelated districts or multiply records.
# Keep the tables separate until a reliable common key is known.

ifi_impact_context = ifi_impact.copy()
ifi_area_context = ifi_area.copy()
ifi_severity_context = ifi_severity.copy()

print("\nIFI tables retained separately:")
print("Impact table:", ifi_impact_context.shape)
print("Flooded-area table:", ifi_area_context.shape)
print("Severity table:", ifi_severity_context.shape)


# ------------------------------------------------------------
# 6. Prepare helper functions
# ------------------------------------------------------------

rng = np.random.default_rng(42)

def sample_row(dataframe):
    """Select one source row for illustrative contextual sampling."""
    if dataframe.empty:
        raise ValueError("A source dataset is empty; cannot sample from it.")
    return dataframe.iloc[int(rng.integers(0, len(dataframe)))]

def safe_number(row, column, default=np.nan):
    """Read a numeric source value safely."""
    if column not in row.index:
        return default

    value = pd.to_numeric(
        pd.Series([row[column]]),
        errors="coerce"
    ).iloc[0]

    return float(value) if pd.notna(value) else default

def yes_no(condition):
    """Convert a condition into a consistent Yes/No value."""
    return "Yes" if bool(condition) else "No"


# Select assistance labels available in the Disaster Response data.
assistance_columns = [
    "medical_help",
    "search_and_rescue",
    "water",
    "food",
    "shelter",
    "clothing",
    "transport",
    "buildings",
    "floods",
]

assistance_columns = [
    column for column in assistance_columns
    if column in disaster_messages.columns
]


# ------------------------------------------------------------
# 7. Define illustrative urgency rules
# ------------------------------------------------------------

def assign_synthetic_urgency(row):
    """
    Example-only rules for generated scenarios.
    They are not a clinically or operationally validated
    emergency triage standard.
    """

    if (
        row["medical_emergency"] == "Yes"
        or (
            row["people_trapped"] == "Yes"
            and row["water_level"] == "Rapidly_Rising"
        )
        or (
            row["building_condition"] == "Collapsing"
            and row["people_trapped"] == "Yes"
        )
    ):
        return "Critical"

    if (
        row["people_trapped"] == "Yes"
        or row["injuries_reported"] == "Yes"
        or row["rescue_required"] == "Yes"
        or row["location_risk"] == "High"
    ):
        return "High"

    if (
        row["location_risk"] == "Medium"
        or row["food_required"] == "Yes"
        or row["water_required"] == "Yes"
        or row["shelter_required"] == "Yes"
        or row["building_condition"] == "Damaged"
    ):
        return "Medium"

    return "Low"


# ------------------------------------------------------------
# 8. Generate additional synthetic SOS scenarios
# ------------------------------------------------------------

N_SYNTHETIC = 5000

water_levels = [
    "Stable",
    "Rising",
    "Rapidly_Rising",
]

building_conditions = [
    "Intact",
    "Damaged",
    "Collapsing",
    "Unknown",
]

location_risks = [
    "Low",
    "Medium",
    "High",
]

synthetic_records = []

for i in range(N_SYNTHETIC):

    # These are independent context samples.
    # They are NOT confirmed to describe the same real location.
    event_row = sample_row(flood_context)
    impact_row = sample_row(ifi_impact_context)
    area_row = sample_row(ifi_area_context)
    severity_row = sample_row(ifi_severity_context)
    message_row = sample_row(disaster_messages)

    # Generate illustrative SOS-level features.
    people_affected = int(rng.integers(1, 31))

    people_trapped = rng.choice(
        ["Yes", "No"],
        p=[0.25, 0.75]
    )

    injuries_reported = rng.choice(
        ["Yes", "No"],
        p=[0.30, 0.70]
    )

    medical_emergency = rng.choice(
        ["Yes", "No"],
        p=[0.18, 0.82]
    )

    water_level = rng.choice(
        water_levels,
        p=[0.30, 0.45, 0.25]
    )

    building_condition = rng.choice(
        building_conditions,
        p=[0.55, 0.25, 0.05, 0.15]
    )

    location_risk = rng.choice(
        location_risks,
        p=[0.25, 0.40, 0.35]
    )

    # Read assistance categories from a sampled Disaster Response row.
    # These categories are contextual examples, not verified needs
    # for the generated SOS scenario.
    assistance_values = {}

    for column in assistance_columns:
        value = pd.to_numeric(
            pd.Series([message_row[column]]),
            errors="coerce"
        ).iloc[0]

        assistance_values[column] = (
            int(value) if pd.notna(value) else np.nan
        )

    # Use SOS features for generated resource requests.
    rescue_required = yes_no(
        people_trapped == "Yes"
        or injuries_reported == "Yes"
        or medical_emergency == "Yes"
    )

    food_required = yes_no(
        rng.random() < 0.30
    )

    water_required = yes_no(
        rng.random() < 0.35
    )

    shelter_required = yes_no(
        building_condition in ["Damaged", "Collapsing"]
        or rng.random() < 0.20
    )

    # Construct the scenario.
    record = {
        "sos_id": f"SYN{i + 1:06d}",

        "sos_description": (
            f"Synthetic flood scenario affecting {people_affected} people. "
            f"Water level: {water_level}. "
            f"Building condition: {building_condition}. "
            f"Location risk: {location_risk}."
        ),

        "people_affected": people_affected,
        "people_trapped": people_trapped,
        "injuries_reported": injuries_reported,
        "medical_emergency": medical_emergency,
        "water_level": water_level,
        "building_condition": building_condition,
        "rescue_required": rescue_required,
        "food_required": food_required,
        "water_required": water_required,
        "shelter_required": shelter_required,
        "location_risk": location_risk,

        # INDOFLOODS context.
        "indoflood_event_id_context": event_row.get(
            "EventID", np.nan
        ),
        "context_peak_flood_level_m": safe_number(
            event_row, "Peak Flood Level (m)"
        ),
        "context_event_duration_days": safe_number(
            event_row, "Event Duration (days)"
        ),
        "context_rainfall_t1d": safe_number(
            event_row, "T1d"
        ),

        # IFI context.
        "ifi_impact_district_context": impact_row.get(
            "Dist_Name", np.nan
        ),
        "context_district_fatalities": safe_number(
            impact_row, "Human_fatality"
        ),
        "context_district_injuries": safe_number(
            impact_row, "Human_injured"
        ),
        "context_district_population": safe_number(
            impact_row, "Population"
        ),

        "ifi_area_district_context": area_row.get(
            "Dist_Name", np.nan
        ),
        "context_percent_flooded_area": safe_number(
            area_row, "Corrected_Percent_Flooded_Area",
            default=safe_number(area_row, "Percent_Flooded_Area")
        ),

        "ifi_severity_district_context": severity_row.get(
            "District_Name", np.nan
        ),
        "context_dfsi": safe_number(severity_row, "DFSI"),

        # Disaster Response context.
        "disaster_response_message_id_context": message_row.get(
            "id", np.nan
        ),
        "disaster_response_message_context": str(
            message_row["message"]
        ),

        # Retain available assistance label columns with a source_ prefix.
        **{
            f"source_{column}": value
            for column, value in assistance_values.items()
        },

        # Provenance.
        "record_source": "synthetic",
        "label_source": "synthetic_rule",
        "context_link_status": "independent_sample_not_linked",
    }

    record["urgency_label"] = assign_synthetic_urgency(record)
    synthetic_records.append(record)

synthetic_df = pd.DataFrame(synthetic_records)

print("\nSynthetic SOS scenarios created:", len(synthetic_df))


# ------------------------------------------------------------
# 9. Prepare the original SOS records
# ------------------------------------------------------------

# Preserve original SOS records and existing labels.
# Their labels have not been independently verified.
original_df = sos_original.copy()

if "record_source" not in original_df.columns:
    original_df["record_source"] = "unknown"

if "label_source" not in original_df.columns:
    original_df["label_source"] = "existing_label_unverified"

if "context_link_status" not in original_df.columns:
    original_df["context_link_status"] = "not_recorded"

# Add any additional generated-dataset columns to original records
# as missing, since no reliable contextual linkage has been confirmed.
for column in synthetic_df.columns:
    if column not in original_df.columns:
        original_df[column] = np.nan

for column in original_df.columns:
    if column not in synthetic_df.columns:
        synthetic_df[column] = np.nan

# Use the same column order in both dataframes.
synthetic_df = synthetic_df[original_df.columns]

# Combine original records first, followed by synthetic scenarios.
final_df = pd.concat(
    [original_df, synthetic_df],
    ignore_index=True
)

# Remove duplicate IDs only; retain the first occurrence.
# Synthetic IDs use a separate SYN prefix.
final_df = final_df.drop_duplicates(
    subset=["sos_id"],
    keep="first"
).reset_index(drop=True)


# ------------------------------------------------------------
# 10. Validate the final dataset
# ------------------------------------------------------------

valid_labels = [
    "Low",
    "Medium",
    "High",
    "Critical",
]

print("\n" + "=" * 50)
print("FINAL SOS URGENCY DATASET REPORT")
print("=" * 50)

print("Total rows:", len(final_df))
print("Total columns:", len(final_df.columns))

print(
    "Duplicate SOS IDs:",
    final_df["sos_id"].duplicated().sum()
)

print(
    "Missing SOS IDs:",
    final_df["sos_id"].isna().sum()
)

print(
    "Invalid urgency labels:",
    (~final_df["urgency_label"].isin(valid_labels)).sum()
)

print("\nRecords by source:")
display(
    final_df["record_source"].value_counts(dropna=False)
)

print("\nLabel-source distribution:")
display(
    final_df["label_source"].value_counts(dropna=False)
)

print("\nUrgency-label distribution:")
display(
    final_df["urgency_label"]
    .value_counts(dropna=False)
    .reindex(valid_labels, fill_value=0)
)

print("\nMissing values in core columns:")
core_columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "urgency_label",
    "record_source",
    "label_source",
]

display(
    final_df[core_columns]
    .isna()
    .sum()
    .to_frame("Missing_Count")
)

print("\nDataset preview:")
display(final_df.head(10))


# ------------------------------------------------------------
# 11. Save the processed dataset
# ------------------------------------------------------------

# Save a new output file. The collection CSV is not overwritten.
final_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\nDataset successfully saved to:")
print(OUTPUT_FILE.resolve())

print("\nOriginal collection file was not overwritten.")
print("Model training has NOT started.")
```

```python

# STEP 5: Final preprocessing of the S
S urgency dataset
# This cell only cleans and prepares the existing dataset.
# It does not train a model or create additional SOS scenarios.

from pathlib import Path
import pandas as pd

# --------------------------------------------------
# 1. Load the existing processed dataset
# --------------------------------------------------

input_file = Path.cwd() / "sos_urgency_dataset_processed.csv"

if not input_file.exists():
    raise FileNotFoundError(
        f"Dataset not found: {input_file.resolve()}"
    )

df = pd.read_csv(input_file, low_memory=False)

print("Dataset loaded:", df.shape)


# --------------------------------------------------
# 2. Remove duplicate SOS IDs
# --------------------------------------------------

# Keep the first occurrence of each SOS ID.
before = len(df)

df = df.drop_duplicates(
    subset=["sos_id"],
    keep="first"
).copy()

print("Duplicate-ID rows removed:", before - len(df))


# --------------------------------------------------
# 3. Clean text columns
# --------------------------------------------------

# Remove leading/trailing spaces from text values.
# Do not change the capitalization of descriptive text.
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    df[column] = df[column].apply(
        lambda value: value.strip()
        if isinstance(value, str)
        else value
    )

# Convert empty text strings to missing values.
df[text_columns] = df[text_columns].replace(
    r"^\s*$",
    pd.NA,
    regex=True
)


# --------------------------------------------------
# 4. Standardize key categorical columns
# --------------------------------------------------

# Standardize Yes/No values in SOS features.
yes_no_columns = [
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
]

for column in yes_no_columns:
    if column in df.columns:
        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
            .str.title()
        )

# Standardize urgency labels.
if "urgency_label" in df.columns:
    df["urgency_label"] = (
        df["urgency_label"]
        .astype("string")
        .str.strip()
        .str.title()
    )

# Standardize other categorical SOS features.
categorical_columns = [
    "water_level",
    "building_condition",
    "location_risk",
]

for column in categorical_columns:
    if column in df.columns:
        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
        )


# --------------------------------------------------
# 5. Check required fields and invalid labels
# --------------------------------------------------

required_columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "urgency_label",
]

missing_required_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_required_columns:
    raise ValueError(
        f"Required columns are missing: {missing_required_columns}"
    )

valid_labels = ["Low", "Medium", "High", "Critical"]

# Keep records with a known urgency label and SOS ID.
before = len(df)

df = df.dropna(
    subset=["sos_id", "urgency_label"]
).copy()

df = df[
    df["urgency_label"].isin(valid_labels)
].copy()

print("Rows removed for missing IDs/labels or invalid labels:",
      before - len(df))


# --------------------------------------------------
# 6. Convert numeric features to numeric types
# --------------------------------------------------

numeric_columns = [
    "people_affected",
    "context_peak_flood_level_m",
    "context_event_duration_days",
    "context_rainfall_t1d",
    "context_district_fatalities",
    "context_district_injuries",
    "context_district_population",
    "context_percent_flooded_area",
    "context_dfsi",
]

for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        )

# Do not invent replacement values for missing numeric data.
# Keep missing contextual values as missing until feature selection.


# --------------------------------------------------
# 7. Check impossible people counts
# --------------------------------------------------

if "people_affected" in df.columns:
    invalid_count = (
        df["people_affected"].notna()
        & (
            (df["people_affected"] < 1)
            | (df["people_affected"] % 1 != 0)
        )
    )

    print("Rows with invalid people_affected values:",
          int(invalid_count.sum()))

    # Set invalid values to missing rather than inventing a count.
    df.loc[invalid_count, "people_affected"] = pd.NA


# --------------------------------------------------
# 8. Add preprocessing provenance
# --------------------------------------------------

df["preprocessing_status"] = "cleaned"

# Preserve existing record_source and label_source columns.
# Synthetic records must remain identifiable as synthetic.


# --------------------------------------------------
# 9. Save the final cleaned dataset
# --------------------------------------------------

output_file = Path.cwd() / "sos_urgency_dataset_preprocessed.csv"

df.to_csv(output_file, index=False)

print("\nFINAL PREPROCESSING REPORT")
print("--------------------------")
print("Final rows:", len(df))
print("Final columns:", len(df.columns))
print("Duplicate SOS IDs:", df["sos_id"].duplicated().sum())
print("Missing SOS IDs:", df["sos_id"].isna().sum())
print("Invalid urgency labels:",
      (~df["urgency_label"].isin(valid_labels)).sum())

print("\nUrgency label distribution:")
display(
    df["urgency_label"]
    .value_counts()
    .reindex(valid_labels, fill_value=0)
)

print("\nMissing values in core columns:")
display(
    df[
        ["sos_id", "sos_description", "people_affected", "urgency_label"]
    ].isna().sum().to_frame("Missing_Count")
)

print("\nSaved cleaned dataset to:")
print(output_file.resolve())
```

```python

# STEP 7: Encode SOS features for future model training
# This is preprocessing only. No model is trained.

from pathlib import Path
import pandas as pd

# Load the cleaned dataset
input_file = Path.cwd() / "sos_urgency_dataset_preprocessed.csv"

if not input_file.exists():
    raise FileNotFoundError(
        "Run the previous preprocessing steps first."
    )

df = pd.read_csv(input_file, low_memory=False)

# Define the SOS features and prediction target
feature_columns = [
    "people_affected",
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "water_level",
    "building_condition",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
    "location_risk",
]

target_column = "urgency_label"

# Confirm required columns exist
required_columns = feature_columns + [target_column]
missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# Make a copy so the cleaned dataset stays unchanged
model_data = df[required_columns].copy()

# Convert Yes/No features to 1/0
binary_columns = [
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
]

for column in binary_columns:
    model_data[column] = (
        model_data[column]
        .astype("string")
        .str.strip()
        .str.title()
        .map({"Yes": 1, "No": 0})
    )

# Convert people_affected to numeric
model_data["people_affected"] = pd.to_numeric(
    model_data["people_affected"],
    errors="coerce"
)

# One-hot encode the remaining categorical features.
categorical_columns = [
    "water_level",
    "building_condition",
    "location_risk",
]

model_data = pd.get_dummies(
    model_data,
    columns=categorical_columns,
    dummy_na=True,
    dtype=int
)

# Handle missing numeric feature values using the median.
# If a column is entirely missing, fill it with zero.
feature_names = [
    column for column in model_data.columns
    if column != target_column
]

for column in feature_names:
    model_data[column] = pd.to_numeric(
        model_data[column],
        errors="coerce"
    )

    median_value = model_data[column].median()

    if pd.isna(median_value):
        median_value = 0

    model_data[column] = model_data[column].fillna(median_value)

# Remove records without a valid target label.
valid_labels = ["Low", "Medium", "High", "Critical"]

before = len(model_data)

model_data = model_data[
    model_data[target_column].isin(valid_labels)
].copy()

print("Rows removed due to invalid/missing target:",
      before - len(model_data))

# Save the encoded dataset separately.
output_file = Path.cwd() / "sos_urgency_dataset_encoded.csv"

model_data.to_csv(output_file, index=False)

print("\nEncoded dataset shape:", model_data.shape)
print("Remaining missing feature values:",
      int(model_data[feature_names].isna().sum().sum()))

print("\nUrgency label distribution:")
display(model_data[target_column].value_counts())

print("\nEncoded feature preview:")
display(model_data.head())

print("\nSaved to:")
print(output_file.resolve())

print("\nNo model training has been performed.")
```

```python

# JEEVAN LINK: FINAL SOS URGENCY DATASET CLEANING
# Uses the actual CSV filename from your preprocessing summary.
# This cell performs preprocessing only; it does not train a model.

from pathlib import Path
import pandas as pd
import numpy as np

# --------------------------------------------------
# 1. Load the existing encoded dataset
# --------------------------------------------------

base_dir = Path.cwd()

input_file = base_dir / "sos_urgency_dataset_final_preproccessedfile.csv"

if not input_file.exists():
    raise FileNotFoundError(
        f"File not found: {input_file.resolve()}"
    )

df = pd.read_csv(input_file, low_memory=False)

print("Input dataset shape:", df.shape)


# --------------------------------------------------
# 2. Validate the target column
# --------------------------------------------------

target_column = "urgency_label"

if target_column not in df.columns:
    raise ValueError(
        f"Target column '{target_column}' is missing. "
        f"Available columns: {df.columns.tolist()}"
    )

valid_labels = ["Low", "Medium", "High", "Critical"]

df[target_column] = (
    df[target_column]
    .astype("string")
    .str.strip()
    .str.title()
)

before = len(df)

df = df[
    df[target_column].isin(valid_labels)
].copy()

invalid_target_rows_removed = before - len(df)


# --------------------------------------------------
# 3. Remove exact duplicate feature/label rows
# --------------------------------------------------

before = len(df)

comparison_columns = list(df.columns)

df = df.drop_duplicates(
    subset=comparison_columns,
    keep="first"
).reset_index(drop=True)

duplicate_rows_removed = before - len(df)


# --------------------------------------------------
# 4. Remove constant feature columns
# --------------------------------------------------

feature_columns = [
    column for column in df.columns
    if column != target_column
]

constant_columns = [
    column for column in feature_columns
    if df[column].nunique(dropna=False) <= 1
]

df = df.drop(columns=constant_columns)


# --------------------------------------------------
# 5. Validate numeric features
# --------------------------------------------------

feature_columns = [
    column for column in df.columns
    if column != target_column
]

# The uploaded file is expected to contain encoded features.
# If any remaining feature is non-numeric, report it instead
# of silently converting all its values to missing.
non_numeric_columns = [
    column for column in feature_columns
    if not pd.api.types.is_numeric_dtype(df[column])
]

if non_numeric_columns:
    raise ValueError(
        "These features are not numeric. Inspect their values "
        "before encoding them:\n"
        + "\n".join(non_numeric_columns)
    )

# Report missing values rather than inventing replacement values.
missing_by_column = df[feature_columns].isna().sum()
missing_feature_values = int(missing_by_column.sum())

print("\nMissing values by feature:")
display(missing_by_column[missing_by_column > 0].to_frame("Missing_Count"))


# --------------------------------------------------
# 6. Validate labels and final dataset
# --------------------------------------------------

invalid_labels_remaining = int(
    (~df[target_column].isin(valid_labels)).sum()
)

print("\nFinal urgency label counts:")
display(
    df[target_column]
    .value_counts()
    .reindex(valid_labels, fill_value=0)
)

print("\nRows removed for invalid labels:", invalid_target_rows_removed)
print("Exact duplicate rows removed:", duplicate_rows_removed)
print("Constant columns removed:", constant_columns)
print("Remaining missing feature values:", missing_feature_values)
print("Invalid labels remaining:", invalid_labels_remaining)


# --------------------------------------------------
# 7. Save a new final cleaned CSV
# --------------------------------------------------

output_file = base_dir / "sos_urgency_dataset_final_cleaned.csv"

df.to_csv(output_file, index=False)

# Save a cleaning report alongside the dataset.
report_df = pd.DataFrame({
    "Check": [
        "Input rows",
        "Final rows",
        "Final columns",
        "Feature count",
        "Invalid target rows removed",
        "Duplicate rows removed",
        "Constant columns removed",
        "Missing feature values remaining",
        "Invalid labels remaining",
    ],
    "Result": [
        len(pd.read_csv(input_file, low_memory=False)),
        len(df),
        len(df.columns),
        len(feature_columns),
        invalid_target_rows_removed,
        duplicate_rows_removed,
        len(constant_columns),
        missing_feature_values,
        invalid_labels_remaining,
    ],
})

report_file = base_dir / "sos_urgency_final_cleaning_report.csv"
report_df.to_csv(report_file, index=False)

print("\nFINAL CLEANED DATASET")
print("Shape:", df.shape)
print("Saved to:", output_file.resolve())
print("Cleaning report:", report_file.resolve())
print("\nPreprocessing only — no model was trained.")
```

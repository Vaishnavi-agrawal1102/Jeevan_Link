```python

# ============================================================
# STEP 10: PREPROCESS THE SECOND IFI FILE
# File: District_FloodImpact.csv
# ============================================================

import pandas as pd
from pathlib import Path

# 1. Define the input and output paths.
raw_path = Path("../../Raw_Data/ifi_Data/District_FloodImpact.csv")
output_path = Path("District_FloodImpact_processed.csv")

# 2. Load the original district flood-impact dataset.
df_impact = pd.read_csv(raw_path, low_memory=False)

# 3. Display the original dataset information.
print("Original dataset shape:", df_impact.shape)
print("\nOriginal column names:")
print(df_impact.columns.tolist())

# 4. Create a working copy so the raw data stays unchanged.
df_impact_processed = df_impact.copy()

# 5. Remove extra spaces from column names.
df_impact_processed.columns = (
    df_impact_processed.columns.astype(str).str.strip()
)

# 6. Identify and remove columns that are completely empty.
empty_columns = df_impact_processed.columns[
    df_impact_processed.isna().all()
].tolist()

df_impact_processed = df_impact_processed.drop(
    columns=empty_columns
)

print("\nCompletely empty columns removed:", empty_columns)

# 7. Remove exact duplicate rows.
duplicates_removed = int(
    df_impact_processed.duplicated().sum()
)

df_impact_processed = (
    df_impact_processed.drop_duplicates().copy()
)

print("Exact duplicate rows removed:", duplicates_removed)

# 8. Clean extra spaces from text values.
# Missing values are kept missing rather than replaced with invented data.
for column in df_impact_processed.select_dtypes(
    include=["object", "string"]
).columns:
    df_impact_processed[column] = (
        df_impact_processed[column].map(
            lambda value: value.strip()
            if isinstance(value, str)
            else value
        )
    )

# 9. Display the final columns and remaining missing values.
print("\nProcessed dataset shape:", df_impact_processed.shape)

print("\nProcessed column names:")
print(df_impact_processed.columns.tolist())

print("\nMissing values per column:")
print(
    df_impact_processed.isna()
    .sum()
    .sort_values(ascending=False)
)

# 10. Save the processed dataset.
df_impact_processed.to_csv(output_path, index=False)

print("\nProcessed file saved to:")
print(output_path.resolve())

# 11. Save a simple preprocessing report.
report_path = Path("District_FloodImpact_preprocessing_report.txt")

with open(report_path, "w", encoding="utf-8") as report:
    report.write("DISTRICT FLOOD IMPACT PREPROCESSING REPORT\n")
    report.write(f"Original shape: {df_impact.shape}\n")
    report.write(f"Empty columns removed: {empty_columns}\n")
    report.write(f"Duplicate rows removed: {duplicates_removed}\n")
    report.write(
        f"Processed shape: {df_impact_processed.shape}\n"
    )
    report.write("\nRemaining missing values:\n")
    report.write(
        df_impact_processed.isna()
        .sum()
        .sort_values(ascending=False)
        .to_string()
    )

print("Report saved to:")
print(report_path.resolve())
```

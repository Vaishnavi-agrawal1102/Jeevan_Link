```python

# STEP 12: PREPROCESS THE DISTRICT FLOODED AREA DATASET

import pandas as pd
from pathlib import Path

# 1. Define the input and output file paths.
raw_path = Path("../../Raw_Data/ifi_Data/District_FloodedArea.csv")
output_path = Path("District_FloodedArea_processed.csv")

# 2. Load the original dataset.
df_area = pd.read_csv(raw_path)

# 3. Create a separate working copy to protect the raw data.
df_area_processed = df_area.copy()

# 4. Clean column names by removing extra spaces.
df_area_processed.columns = (
    df_area_processed.columns.astype(str).str.strip()
)

# 5. Remove completely empty columns, if any exist.
empty_columns = df_area_processed.columns[
    df_area_processed.isna().all()
].tolist()

df_area_processed = df_area_processed.drop(
    columns=empty_columns
)

# 6. Remove exact duplicate rows.
duplicates_removed = int(
    df_area_processed.duplicated().sum()
)

df_area_processed = (
    df_area_processed.drop_duplicates().copy()
)

# 7. Trim unnecessary spaces from district names.
if "Dist_Name" in df_area_processed.columns:
    df_area_processed["Dist_Name"] = (
        df_area_processed["Dist_Name"].astype("string").str.strip()
    )

# 8. Convert the flooded-area measurements to numeric values.
numeric_columns = [
    "Percent_Flooded_Area",
    "Parmanent_Water",
    "Corrected_Percent_Flooded_Area"
]

for column in numeric_columns:
    if column in df_area_processed.columns:
        df_area_processed[column] = pd.to_numeric(
            df_area_processed[column],
            errors="coerce"
        )

# 9. Display the preprocessing results.
print("Original dataset shape:", df_area.shape)
print("Empty columns removed:", empty_columns)
print("Duplicate rows removed:", duplicates_removed)
print("Processed dataset shape:", df_area_processed.shape)

print("\nMissing values after processing:")
print(df_area_processed.isna().sum())

# 10. Save the processed dataset.
df_area_processed.to_csv(output_path, index=False)

print("\nProcessed dataset saved to:")
print(output_path.resolve())

# 11. Save a preprocessing report.
report_path = Path("District_FloodedArea_preprocessing_report.txt")

with open(report_path, "w", encoding="utf-8") as report:
    report.write("DISTRICT FLOODED AREA PREPROCESSING REPORT\n")
    report.write(f"Original shape: {df_area.shape}\n")
    report.write(f"Empty columns removed: {empty_columns}\n")
    report.write(f"Duplicate rows removed: {duplicates_removed}\n")
    report.write(f"Processed shape: {df_area_processed.shape}\n")
    report.write("\nMissing values after processing:\n")
    report.write(df_area_processed.isna().sum().to_string())

print("Preprocessing report saved to:")
print(report_path.resolve())
```

```python


# STEP 2: LOAD THE RAW IFI DATASET
# Import pandas to read and inspect CSV files.
import pandas as pd

# Specify the exact path to the original IFI CSV file.
raw_path = "../../Raw_Data/ifi_Data/India_Flood_Inventory_v3 (1).csv"

# Load the raw dataset into a DataFrame.
df = pd.read_csv(raw_path)

# Display the number of rows and columns.
print("Dataset shape:", df.shape)

# Display the first five rows.
print(df.head())

```

```python

# STEP 3: CREATE A WORKING COPY
# Keep the original DataFrame unchanged.
# All preprocessing will be performed on df_work.

df_work = df.copy()

# Confirm that the working copy has the same dimensions as the original.
print("Raw dataset shape:", df.shape)
print("Working dataset shape:", df_work.shape)

```

```python

# STEP 4: INSPECT COLUMN NAMES
# Display all column names in the working dataset.
# We need to understand the available information before preprocessing.

print("Total columns:", len(df_work.columns))
print("\nColumn names:")
print(df_work.columns.tolist())
```

```python

# STEP 5: INSPECT DATA TYPES AND NON-MISSING VALUES
# Check each column's data type and how many values are present.
# This helps us identify columns that may need conversion or cleaning.

df_work.info()
```

```python

# STEP 6: CHECK FOR DUPLICATE ROWS
# Count rows that are exact duplicates across all columns.
# We will inspect the result before deciding whether to remove anything.

duplicate_count = df_work.duplicated().sum()

print("Total rows:", len(df_work))
print("Exact duplicate rows:", duplicate_count)
```

```python

# STEP 7: CHECK MISSING VALUES
# Count missing values in every column of the working dataset.
# Display only columns that contain at least one missing value.

missing_values = df_work.isnull().sum()

print("Columns with missing values:")
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

```python

# STEP 7: CHECK MISSING VALUES
# Count missing values in each column of the IFI working dataset.
# We will use this information to decide what needs cleaning.

missing_values = df_work.isnull().sum()

# Show only columns that contain missing values.
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

```python

# STEP 8: INSPECT THE DATA
# Display a few rows to see how values appear in the original columns.
# This helps us investigate why several columns are completely missing.

print(df_work.head().to_string())

# Display the data types of all columns.
print("\nColumn data types:")
print(df_work.dtypes)
```

```python

# STEP 9: FIND COMPLETELY EMPTY COLUMNS
# Count non-missing values in each column.
# Columns with zero non-missing values contain no usable data in this file.

non_missing_counts = df_work.notna().sum()

# Display columns that have no non-missing values.
empty_columns = non_missing_counts[non_missing_counts == 0]

print("Completely empty columns:")
print(empty_columns.index.tolist())
```

```python

# STEP 9: COMPLETE BASIC PREPROCESSING FOR THE IFI DATASET

import pandas as pd
from pathlib import Path

# 1. Create a fresh working copy so the original df remains unchanged.
df_processed = df.copy()

# 2. Record the original size and identify columns that are entirely empty.
original_shape = df_processed.shape
empty_columns = df_processed.columns[df_processed.isna().all()].tolist()

print("Original dataset shape:", original_shape)
print("Completely empty columns:", empty_columns)

# 3. Remove completely empty columns because they contain no usable values.
df_processed = df_processed.drop(columns=empty_columns)

# 4. Remove exact duplicate rows.
duplicate_count = int(df_processed.duplicated().sum())
df_processed = df_processed.drop_duplicates().copy()

# 5. Remove extra spaces from column names.
df_processed.columns = df_processed.columns.astype(str).str.strip()

# 6. Convert the date columns into proper date values.
# Dates that cannot be interpreted will become missing, not invented.
for col in ["Start Date", "End Date"]:
    if col in df_processed.columns:
        df_processed[col] = pd.to_datetime(
            df_processed[col],
            errors="coerce",
            dayfirst=True
        )

# 7. Convert relevant measurement columns to numeric values.
# Invalid values become NaN so they can be reviewed later.
numeric_columns = [
    "Duration(Days)",
    "Human fatality",
    "Human injured",
    "Human Displaced",
    "Animal Fatality",
    "Latitude",
    "Longitude",
    "Area Affected"
]

for col in numeric_columns:
    if col in df_processed.columns:
        df_processed[col] = pd.to_numeric(
            df_processed[col],
            errors="coerce"
        )

# 8. Trim extra spaces from text values without changing missing values.
for col in df_processed.select_dtypes(include=["object", "string"]).columns:
    df_processed[col] = df_processed[col].map(
        lambda value: value.strip() if isinstance(value, str) else value
    )

# 9. Display the results of preprocessing.
print("\nExact duplicate rows removed:", duplicate_count)
print("Processed dataset shape:", df_processed.shape)
print("\nRemaining missing values:")
print(df_processed.isna().sum().sort_values(ascending=False))

# 10. Save the processed dataset in the current notebook folder.
output_path = Path("IFI_flood_events_processed.csv")
df_processed.to_csv(output_path, index=False)

# 11. Save a report describing the preprocessing changes.
report_path = Path("IFI_preprocessing_report.txt")

with open(report_path, "w", encoding="utf-8") as report:
    report.write("IFI FLOOD DATASET PREPROCESSING REPORT\n")
    report.write(f"Original shape: {original_shape}\n")
    report.write(f"Completely empty columns removed: {empty_columns}\n")
    report.write(f"Exact duplicate rows removed: {duplicate_count}\n")
    report.write(f"Processed shape: {df_processed.shape}\n")
    report.write(f"Processed CSV: {output_path.resolve()}\n")
    report.write("\nRemaining missing values:\n")
    report.write(df_processed.isna().sum().sort_values(ascending=False).to_string())

# 12. Confirm where the output files were saved.
print("\nProcessed dataset saved to:", output_path.resolve())
print("Preprocessing report saved to:", report_path.resolve())
```

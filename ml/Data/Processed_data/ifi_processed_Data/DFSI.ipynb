```python

# STEP 14: PREPROCESS THE DFSI DATASET

import pandas as pd
from pathlib import Path

# 1. Define the input and output paths.
raw_path = Path("../../Raw_Data/ifi_Data/DFSI.csv")
output_path = Path("DFSI_processed.csv")

# 2. Load the original DFSI dataset.
df_dfsi = pd.read_csv(raw_path, low_memory=False)

# 3. Create a separate working copy.
df_dfsi_processed = df_dfsi.copy()

# 4. Remove extra spaces from column names.
df_dfsi_processed.columns = (
    df_dfsi_processed.columns.astype(str).str.strip()
)

# 5. Rename columns to describe their apparent meanings.
# The first column contains district names in the sample you showed.
df_dfsi_processed = df_dfsi_processed.rename(
    columns={
        "Unnamed: 0": "District_Name",
        "State_Name": "State_Name",
        "DFSI": "DFSI"
    }
)

# 6. Clean district and state names.
for column in ["District_Name", "State_Name"]:
    if column in df_dfsi_processed.columns:
        df_dfsi_processed[column] = (
            df_dfsi_processed[column]
            .astype("string")
            .str.strip()
        )

# 7. Convert DFSI values to numeric.
# Invalid values become missing rather than being guessed.
df_dfsi_processed["DFSI"] = pd.to_numeric(
    df_dfsi_processed["DFSI"],
    errors="coerce"
)

# 8. Remove exact duplicate rows.
duplicates_removed = int(
    df_dfsi_processed.duplicated().sum()
)
df_dfsi_processed = (
    df_dfsi_processed.drop_duplicates().copy()
)

# 9. Inspect missing values and DFSI statistics.
print("Original dataset shape:", df_dfsi.shape)
print("Duplicate rows removed:", duplicates_removed)
print("Processed dataset shape:", df_dfsi_processed.shape)

print("\nMissing values:")
print(df_dfsi_processed.isna().sum())

print("\nDFSI statistics:")
print(df_dfsi_processed["DFSI"].describe())

# 10. Save the processed dataset.
df_dfsi_processed.to_csv(output_path, index=False)
print("\nProcessed dataset saved to:", output_path.resolve())

# 11. Save a preprocessing report.
report_path = Path("DFSI_preprocessing_report.txt")

with open(report_path, "w", encoding="utf-8") as report:
    report.write("DFSI PREPROCESSING REPORT\n")
    report.write(f"Original shape: {df_dfsi.shape}\n")
    report.write(f"Duplicate rows removed: {duplicates_removed}\n")
    report.write(f"Processed shape: {df_dfsi_processed.shape}\n")
    report.write("\nMissing values:\n")
    report.write(df_dfsi_processed.isna().sum().to_string())
    report.write("\n\nDFSI statistics:\n")
    report.write(df_dfsi_processed["DFSI"].describe().to_string())

print("Report saved to:", report_path.resolve())
```

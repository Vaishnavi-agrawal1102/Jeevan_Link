```python

from pathlib import Path
import pandas as pd

# --------------------------------------------------
# 1. Define the existing data folders
# --------------------------------------------------

BASE_DIR = Path(
    r"C:\Users\ishit\jeevan_link_ml\Data\Processed_data"
)

DISASTER_DIR = BASE_DIR / "Disaster Response Messages"
IFI_DIR = BASE_DIR / "ifi_processed_Data"
INDO_DIR = BASE_DIR / "indoflood_processed_Data"

# Save ONE final CSV directly inside this folder
OUTPUT_DIR = BASE_DIR / "Assitance_Dataset_model"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# --------------------------------------------------
# 2. Select relevant columns from Disaster Response
# --------------------------------------------------

disaster_columns = [
    "message",
    "genre",
    "food",
    "water",
    "shelter",
    "medical_help",
    "medical_products",
    "search_and_rescue",
    "transport",
]

disaster_frames = []

for filename in [
    "disaster_response_training_processed.csv",
    "disaster_response_validation_processed.csv",
    "disaster_response_test_processed.csv",
]:
    file_path = DISASTER_DIR / filename

    df = pd.read_csv(
        file_path,
        usecols=lambda col: col in disaster_columns
    )

    # Identify the kind of record in the master dataset
    df["record_type"] = "disaster_message"

    # Store only the selected columns plus record_type
    df = df[disaster_columns + ["record_type"]]

    disaster_frames.append(df)

disaster_data = pd.concat(
    disaster_frames,
    ignore_index=True
)

print("Disaster Response records:", len(disaster_data))

# --------------------------------------------------
# 3. Select relevant columns from the SOS dataset
# --------------------------------------------------

sos_columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "water_level",
    "building_condition",
    "location_risk",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
    "urgency_label",
]

sos_data = pd.read_csv(
    INDO_DIR / "sos_urgency_dataset_processed.csv",
    usecols=sos_columns
)

sos_data["record_type"] = "sos"

print("SOS records:", len(sos_data))

# --------------------------------------------------
# 4. Combine flood-event and precipitation records
#    using their existing EventID
# --------------------------------------------------

flood_columns = [
    "EventID",
    "Peak Flood Level (m)",
    "Peak Discharge Q (cumec)",
    "Flood Volume (cumec)",
    "Event Duration (days)",
    "Time to Peak (days)",
    "Flood Type",
]

precipitation_columns = [
    "EventID",
    "T1d", "T2d", "T3d", "T4d", "T5d",
    "T6d", "T7d", "T8d", "T9d", "T10d",
]

flood_data = pd.read_csv(
    INDO_DIR / "floodevents_indofloods_processed.csv",
    usecols=flood_columns
)

precipitation_data = pd.read_csv(
    INDO_DIR / "precipitation_variables_indofloods_processed.csv",
    usecols=precipitation_columns
)

# Check that precipitation has no duplicate EventID values
# before joining, to avoid accidentally multiplying flood rows.
if precipitation_data["EventID"].duplicated().any():
    raise ValueError(
        "Duplicate EventID values found in precipitation data. "
        "Inspect them before merging."
    )

flood_data = flood_data.merge(
    precipitation_data,
    on="EventID",
    how="left",
    validate="many_to_one"
)

flood_data["record_type"] = "flood_event"

print("Flood-event records:", len(flood_data))

# --------------------------------------------------
# 5. Give each source its own columns in ONE master CSV
# --------------------------------------------------
# Different record types have different fields.
# Missing fields remain empty rather than inventing values.

master_data = pd.concat(
    [
        disaster_data,
        sos_data,
        flood_data
    ],
    ignore_index=True,
    sort=False
)

# --------------------------------------------------
# 6. Save the single master dataset
# --------------------------------------------------

output_path = OUTPUT_DIR / "assistance_master_dataset.csv"

master_data.to_csv(output_path, index=False)

# --------------------------------------------------
# 7. Verify the output
# --------------------------------------------------

print("\nMASTER DATASET CREATED")
print("Saved at:", output_path)
print("Total rows:", len(master_data))
print("Total columns:", len(master_data.columns))

print("\nRecords by type:")
print(master_data["record_type"].value_counts())

print("\nMaster dataset preview:")
print(master_data.head())

print("\nMissing values by column:")
print(master_data.isna().sum())
```

```python

from pathlib import Path
import pandas as pd

# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\ishit\jeevan_link_ml\Data\Processed_data"
)

DISASTER_DIR = BASE_DIR / "Disaster Response Messages"
IFI_DIR = BASE_DIR / "ifi_processed_Data"
INDO_DIR = BASE_DIR / "indoflood_processed_Data"

# IMPORTANT: This is the correct SOS folder.
SOS_DIR = BASE_DIR / "sos_urgency_dataset_model"

OUTPUT_DIR = BASE_DIR / "Assitance_Dataset_model"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "assistance_master_dataset.csv"


# ============================================================
# 2. HELPER: READ ONLY SELECTED COLUMNS
# ============================================================

def read_selected(folder, filename, columns, record_type):
    """Read relevant columns and identify the source record type."""

    path = folder / filename

    if not path.exists():
        raise FileNotFoundError(f"File not found: {path}")

    # Read only the columns that actually exist in the CSV.
    available = pd.read_csv(path, nrows=0).columns.tolist()
    selected = [column for column in columns if column in available]

    if not selected:
        raise ValueError(
            f"None of the selected columns exist in {path}"
        )

    df = pd.read_csv(path, usecols=selected)

    # Standardize column names to avoid collisions between sources.
    df = df.rename(
        columns={column: f"{record_type}__{column}"
                 for column in selected}
    )

    df.insert(0, "record_type", record_type)

    return df


# ============================================================
# 3. DISASTER RESPONSE MESSAGES
# ============================================================

disaster_columns = [
    "id",
    "message",
    "genre",
    "food",
    "water",
    "shelter",
    "medical_help",
    "medical_products",
    "search_and_rescue",
    "transport",
]

disaster_frames = []

disaster_files = [
    "disaster_response_training_processed.csv",
    "disaster_response_validation_processed.csv",
    "disaster_response_test_processed.csv",
]

for filename in disaster_files:
    df = read_selected(
        DISASTER_DIR,
        filename,
        disaster_columns,
        "disaster_message",
    )

    # Keep the original split as source metadata only.
    # This does not create separate output files.
    source_split = {
        "disaster_response_training_processed.csv": "training",
        "disaster_response_validation_processed.csv": "validation",
        "disaster_response_test_processed.csv": "test",
    }[filename]

    df["source_split"] = source_split
    disaster_frames.append(df)

disaster_data = pd.concat(
    disaster_frames,
    ignore_index=True,
    sort=False,
)

print("Disaster Response rows:", len(disaster_data))


# ============================================================
# 4. SOS DATASET — USE THE CORRECT FILE
# ============================================================

sos_columns = [
    "sos_id",
    "sos_description",
    "people_affected",
    "people_trapped",
    "injuries_reported",
    "medical_emergency",
    "water_level",
    "building_condition",
    "rescue_required",
    "food_required",
    "water_required",
    "shelter_required",
    "location_risk",
    "urgency_label",
    "record_source",
    "label_source",
    "context_link_status",
    "indoflood_event_id_context",
    "context_peak_flood_level_m",
    "context_event_duration_days",
    "context_rainfall_t1d",
    "ifi_impact_district_context",
    "context_district_fatalities",
    "context_district_injuries",
    "context_district_population",
    "ifi_area_district_context",
    "context_percent_flooded_area",
    "ifi_severity_district_context",
    "context_dfsi",
    "disaster_response_message_id_context",
    "disaster_response_message_context",
    "source_medical_help",
    "source_search_and_rescue",
    "source_water",
    "source_food",
    "source_shelter",
    "source_clothing",
    "source_transport",
    "source_buildings",
    "source_floods",
]

sos_data = read_selected(
    SOS_DIR,
    "sos_urgency_dataset_processed.csv",
    sos_columns,
    "sos",
)

print("SOS rows:", len(sos_data))


# ============================================================
# 5. FLOOD EVENTS + PRECIPITATION
# ============================================================

flood_columns = [
    "EventID",
    "Peak Flood Level (m)",
    "Peak Discharge Q (cumec)",
    "Flood Volume (cumec)",
    "Event Duration (days)",
    "Time to Peak (days)",
    "Flood Type",
]

rain_columns = [
    "EventID",
    "T1d", "T2d", "T3d", "T4d", "T5d",
    "T6d", "T7d", "T8d", "T9d", "T10d",
]

flood_path = INDO_DIR / "floodevents_indofloods_processed.csv"
rain_path = INDO_DIR / "precipitation_variables_indofloods_processed.csv"

flood_data = pd.read_csv(
    flood_path,
    usecols=lambda c: c in flood_columns,
)

rain_data = pd.read_csv(
    rain_path,
    usecols=lambda c: c in rain_columns,
)

# Normalize IDs for matching, without changing the source files.
flood_data["EventID"] = flood_data["EventID"].astype("string").str.strip()
rain_data["EventID"] = rain_data["EventID"].astype("string").str.strip()

# Precipitation must have at most one row per EventID for this join.
if rain_data["EventID"].duplicated().any():
    raise ValueError(
        "Precipitation contains duplicate EventID values. "
        "Resolve duplicates before joining."
    )

# Prefix precipitation columns to avoid ambiguous column names.
rain_data = rain_data.rename(
    columns={
        column: f"precipitation__{column}"
        for column in rain_data.columns
        if column != "EventID"
    }
)

flood_data = flood_data.merge(
    rain_data,
    left_on="EventID",
    right_on="EventID",
    how="left",
    validate="many_to_one",
)

# Prefix all flood columns, including EventID.
flood_data = flood_data.rename(
    columns={
        column: f"flood__{column}"
        for column in flood_data.columns
    }
)

flood_data.insert(0, "record_type", "flood_event")

print("Flood-event rows after precipitation join:", len(flood_data))


# ============================================================
# 6. IFI DISTRICT AND EVENT SOURCES
# ============================================================

ifi_sources = [
    (
        "District_FloodImpact_processed.csv",
        [
            "Dist_Name",
            "Human_fatality",
            "Human_injured",
            "Population",
            "Mean_Flood_Duration",
        ],
        "ifi_flood_impact",
    ),
    (
        "District_FloodedArea_processed.csv",
        [
            "Dist_Name",
            "Percent_Flooded_Area",
            "Parmanent_Water",
            "Corrected_Percent_Flooded_Area",
        ],
        "ifi_flooded_area",
    ),
    (
        "DFSI_processed.csv",
        ["District_Name", "State_Name", "DFSI"],
        "ifi_severity",
    ),
    (
        "IFI_flood_events_processed.csv",
        [
            "Start Date",
            "End Date",
            "Duration(Days)",
            "Main Cause",
            "Districts",
            "State",
            "Human Displaced",
            "Extent of damage",
            "District_LGD_Codes",
            "State_Codes",
        ],
        "ifi_flood_event",
    ),
]

ifi_frames = []

for filename, columns, record_type in ifi_sources:
    df = read_selected(
        IFI_DIR,
        filename,
        columns,
        record_type,
    )
    ifi_frames.append(df)
    print(f"{record_type} rows:", len(df))


# ============================================================
# 7. CATCHMENT CHARACTERISTICS
# ============================================================

catchment_columns = [
    "GaugeID",
    "Drainage Area",
    "Catchment Relief",
    "Annual Precipitation",
    "Population Density",
    "Road Density",
    "Urban percentage",
]

catchment_data = read_selected(
    INDO_DIR,
    "catchment_characteristics_indofloods_processed.csv",
    catchment_columns,
    "catchment",
)

print("Catchment rows:", len(catchment_data))


# ============================================================
# 8. COMBINE ALL SOURCES INTO ONE MASTER CSV
# ============================================================
# These sources have different row meanings.
# They are stored in one file with source-specific columns.
# We do not fabricate links between unrelated records.

all_frames = [
    disaster_data,
    sos_data,
    flood_data,
    catchment_data,
] + ifi_frames

master_data = pd.concat(
    all_frames,
    ignore_index=True,
    sort=False,
)

# Remove exact duplicate rows only.
# Do not deduplicate using descriptions or partial matches.
master_data = master_data.drop_duplicates().reset_index(drop=True)

# Save the one final output file.
master_data.to_csv(OUTPUT_FILE, index=False)

# ============================================================
# 9. SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("MASTER ASSISTANCE DATASET CREATED")
print("=" * 60)

print("Output:", OUTPUT_FILE)
print("Total rows:", len(master_data))
print("Total columns:", len(master_data.columns))

print("\nRows by record type:")
print(master_data["record_type"].value_counts(dropna=False))

print("\nFirst five rows:")
print(master_data.head())

print("\nFile size (MB):", round(OUTPUT_FILE.stat().st_size / (1024**2), 2))
```

```python

from pathlib import Path
import pandas as pd

# ============================================================
# 1. PROJECT PATHS
# ============================================================

BASE_DIR = Path(
    r"C:\Users\ishit\jeevan_link_ml\Data\Processed_data"
)

DISASTER_DIR = BASE_DIR / "Disaster Response Messages"
IFI_DIR = BASE_DIR / "ifi_processed_Data"
INDO_DIR = BASE_DIR / "indoflood_processed_Data"
SOS_DIR = BASE_DIR / "sos_urgency_dataset_model"

OUTPUT_DIR = BASE_DIR / "Assitance_Dataset_model"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = OUTPUT_DIR / "assistance_master_dataset.csv"


# ============================================================
# 2. LOAD DISASTER RESPONSE MESSAGES
# ============================================================

disaster_columns = [
    "id", "message", "genre",
    "food", "water", "shelter",
    "medical_help", "medical_products",
    "search_and_rescue", "transport",
]

disaster_frames = []

split_files = {
    "training": "disaster_response_training_processed.csv",
    "validation": "disaster_response_validation_processed.csv",
    "test": "disaster_response_test_processed.csv",
}

for split_name, filename in split_files.items():
    path = DISASTER_DIR / filename

    df = pd.read_csv(
        path,
        usecols=lambda col: col in disaster_columns,
        low_memory=False,
    )

    missing = set(disaster_columns) - set(df.columns)
    if missing:
        raise ValueError(
            f"{filename} is missing columns: {sorted(missing)}"
        )

    df["record_type"] = "disaster_message"
    df["source_split"] = split_name
    disaster_frames.append(df)

disaster = pd.concat(
    disaster_frames,
    ignore_index=True,
    sort=False,
)

print("Disaster Response:", len(disaster))


# ============================================================
# 3. LOAD SOS FROM THE CORRECT FOLDER
# ============================================================

sos_columns = [
    "sos_id", "sos_description",
    "people_affected", "people_trapped",
    "injuries_reported", "medical_emergency",
    "water_level", "building_condition",
    "rescue_required", "food_required",
    "water_required", "shelter_required",
    "location_risk", "urgency_label",
    "record_source", "label_source",
    "context_link_status",
    "indoflood_event_id_context",
    "context_peak_flood_level_m",
    "context_event_duration_days",
    "context_rainfall_t1d",
    "ifi_impact_district_context",
    "context_district_fatalities",
    "context_district_injuries",
    "context_district_population",
    "ifi_area_district_context",
    "context_percent_flooded_area",
    "ifi_severity_district_context",
    "context_dfsi",
    "disaster_response_message_id_context",
    "disaster_response_message_context",
    "source_medical_help", "source_search_and_rescue",
    "source_water", "source_food", "source_shelter",
    "source_clothing", "source_transport",
    "source_buildings", "source_floods",
]

sos_path = SOS_DIR / "sos_urgency_dataset_processed.csv"

sos = pd.read_csv(
    sos_path,
    usecols=lambda col: col in sos_columns,
    low_memory=False,
)

missing = set(sos_columns) - set(sos.columns)
if missing:
    raise ValueError(
        f"SOS dataset is missing columns: {sorted(missing)}"
    )

sos["record_type"] = "sos"

print("SOS:", len(sos))


# ============================================================
# 4. JOIN FLOOD EVENTS TO PRECIPITATION
# ============================================================

flood_columns = [
    "EventID",
    "Peak Flood Level (m)",
    "Peak Discharge Q (cumec)",
    "Flood Volume (cumec)",
    "Event Duration (days)",
    "Time to Peak (days)",
    "Flood Type",
]

rain_columns = [
    "EventID",
    "T1d", "T2d", "T3d", "T4d", "T5d",
    "T6d", "T7d", "T8d", "T9d", "T10d",
]

flood = pd.read_csv(
    INDO_DIR / "floodevents_indofloods_processed.csv",
    usecols=flood_columns,
    low_memory=False,
)

rain = pd.read_csv(
    INDO_DIR / "precipitation_variables_indofloods_processed.csv",
    usecols=rain_columns,
    low_memory=False,
)

flood["EventID"] = flood["EventID"].astype("string").str.strip()
rain["EventID"] = rain["EventID"].astype("string").str.strip()

if rain["EventID"].duplicated().any():
    raise ValueError(
        "Duplicate EventID values found in precipitation data."
    )

rain = rain.rename(
    columns={
        col: f"rain_{col}"
        for col in rain.columns
        if col != "EventID"
    }
)

flood = flood.merge(
    rain,
    on="EventID",
    how="left",
    validate="many_to_one",
)

flood["record_type"] = "flood_event"

print("Flood events:", len(flood))


# ============================================================
# 5. LOAD IFI DATASETS
# ============================================================

ifi_specs = [
    (
        "District_FloodImpact_processed.csv",
        [
            "Dist_Name", "Human_fatality",
            "Human_injured", "Population",
            "Mean_Flood_Duration",
        ],
        "ifi_flood_impact",
    ),
    (
        "District_FloodedArea_processed.csv",
        [
            "Dist_Name", "Percent_Flooded_Area",
            "Parmanent_Water",
            "Corrected_Percent_Flooded_Area",
        ],
        "ifi_flooded_area",
    ),
    (
        "DFSI_processed.csv",
        ["District_Name", "State_Name", "DFSI"],
        "ifi_severity",
    ),
    (
        "IFI_flood_events_processed.csv",
        [
            "Start Date", "End Date", "Duration(Days)",
            "Main Cause", "Districts", "State",
            "Human Displaced", "Extent of damage",
            "District_LGD_Codes", "State_Codes",
        ],
        "ifi_flood_event",
    ),
]

ifi_frames = []

for filename, columns, record_type in ifi_specs:
    path = IFI_DIR / filename

    df = pd.read_csv(
        path,
        usecols=lambda col: col in columns,
        low_memory=False,
    )

    missing = set(columns) - set(df.columns)
    if missing:
        raise ValueError(
            f"{filename} is missing columns: {sorted(missing)}"
        )

    df["record_type"] = record_type
    ifi_frames.append(df)

    print(f"{record_type}:", len(df))


# ============================================================
# 6. LOAD CATCHMENT CHARACTERISTICS
# ============================================================

catchment_columns = [
    "GaugeID", "Drainage Area", "Catchment Relief",
    "Annual Precipitation", "Population Density",
    "Road Density", "Urban percentage",
]

catchment = pd.read_csv(
    INDO_DIR / "catchment_characteristics_indofloods_processed.csv",
    usecols=catchment_columns,
    low_memory=False,
)

catchment["record_type"] = "catchment"

print("Catchment:", len(catchment))


# ============================================================
# 7. STANDARDIZE COLUMN NAMES
# ============================================================
# Prefix columns with their source to avoid name collisions.
# Keep record_type and source_split unchanged.

def standardize_columns(df, prefix):
    rename_map = {
        col: f"{prefix}__{col}"
        for col in df.columns
        if col not in ["record_type", "source_split"]
        and not col.startswith(prefix + "__")
    }
    return df.rename(columns=rename_map)


disaster = standardize_columns(disaster, "disaster_message")
sos = standardize_columns(sos, "sos")
flood = standardize_columns(flood, "flood")
catchment = standardize_columns(catchment, "catchment")

prefixes = [
    "ifi_flood_impact",
    "ifi_flooded_area",
    "ifi_severity",
    "ifi_flood_event",
]

for frame, prefix in zip(ifi_frames, prefixes):
    frame.rename(
        columns={
            col: f"{prefix}__{col}"
            for col in frame.columns
            if col != "record_type"
        },
        inplace=True,
    )


# ============================================================
# 8. COMBINE INTO ONE MASTER CSV
# ============================================================
# Different record types remain separate when there is no
# verified shared identifier. Missing fields stay empty.

all_frames = [
    disaster,
    sos,
    flood,
    catchment,
] + ifi_frames

master = pd.concat(
    all_frames,
    ignore_index=True,
    sort=False,
)

master.to_csv(
    OUTPUT_PATH,
    index=False,
)

# ============================================================
# 9. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("MASTER DATASET SAVED SUCCESSFULLY")
print("=" * 60)

print("Path:", OUTPUT_PATH)
print("Shape:", master.shape)

print("\nRecords by type:")
print(master["record_type"].value_counts())

print("\nExact duplicate rows:", master.duplicated().sum())
print("Completely empty columns:",
      int(master.isna().all().sum()))

print("\nFile size (MB):",
      round(OUTPUT_PATH.stat().st_size / (1024 ** 2), 2))
```

```python

import pandas as pd
from pathlib import Path

# ============================================
# 1. SET THE CORRECT FOLDERS
# ============================================

base_path = Path(
    r"C:\Users\ishit\jeevan_link_ml\Data\Processed_data"
)

disaster_path = base_path / "Disaster Response Messages"

output_path = (
    base_path
    / "Assitance_Dataset_model"
    / "assistance_model_dataset.csv"
)

# ============================================
# 2. FIND CSV FILES AUTOMATICALLY
# ============================================

if not disaster_path.exists():
    raise FileNotFoundError(
        f"Folder does not exist: {disaster_path}"
    )

# Search in this folder and all its subfolders.
csv_files = list(disaster_path.rglob("*.csv"))

print("CSV files found:")
for file in csv_files:
    print(file)

if not csv_files:
    raise FileNotFoundError(
        "No CSV files found inside Disaster Response Messages. "
        "Check the folder path printed above."
    )

# ============================================
# 3. LOAD THE DISASTER MESSAGE FILES
# ============================================

datasets = []

for file_path in csv_files:
    try:
        df = pd.read_csv(file_path)

        # Load only files containing message text and
        # at least one assistance-related label.
        if "message" not in df.columns:
            continue

        label_columns = [
            "food",
            "water",
            "shelter",
            "medical_help",
            "medical_products",
            "search_and_rescue",
            "transport"
        ]

        available_labels = [
            col for col in label_columns
            if col in df.columns
        ]

        if not available_labels:
            continue

        # Identify the source file for each record.
        df["source_file"] = file_path.name
        datasets.append(df)

    except Exception as error:
        print(f"Skipped {file_path.name}: {error}")

if not datasets:
    raise ValueError(
        "CSV files were found, but none contained both "
        "message text and assistance labels. "
        "Check the printed filenames and column names."
    )

disaster_df = pd.concat(datasets, ignore_index=True)

# ============================================
# 4. SELECT ASSISTANCE-RELATED COLUMNS
# ============================================

selected_columns = [
    "id",
    "message",
    "genre",
    "food",
    "water",
    "shelter",
    "medical_help",
    "medical_products",
    "search_and_rescue",
    "transport",
    "source_file"
]

selected_columns = [
    col for col in selected_columns
    if col in disaster_df.columns
]

assistance_df = disaster_df[selected_columns].copy()

# ============================================
# 5. CLEAN THE DATA
# ============================================

if "message" in assistance_df.columns:
    assistance_df["message"] = (
        assistance_df["message"]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    assistance_df = assistance_df[
        assistance_df["message"] != ""
    ]

# Remove duplicate messages by ID, if available.
if "id" in assistance_df.columns:
    assistance_df = assistance_df.drop_duplicates(
        subset=["id"]
    )
else:
    assistance_df = assistance_df.drop_duplicates(
        subset=["message"]
    )

# Convert labels to numeric values.
label_columns = [
    "food",
    "water",
    "shelter",
    "medical_help",
    "medical_products",
    "search_and_rescue",
    "transport"
]

for col in label_columns:
    if col in assistance_df.columns:
        assistance_df[col] = pd.to_numeric(
            assistance_df[col],
            errors="coerce"
        )

# ============================================
# 6. SAVE THE ASSISTANCE DATASET
# ============================================

output_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

assistance_df.to_csv(output_path, index=False)

# ============================================
# 7. DISPLAY THE RESULTS
# ============================================

print("\nAssistance dataset created successfully!")
print("Saved to:", output_path)
print("Dataset shape:", assistance_df.shape)

print("\nColumns:")
print(assistance_df.columns.tolist())

print("\nMissing values:")
print(assistance_df.isna().sum())

print("\nFirst five records:")
print(assistance_df.head())
```

```python

import pandas as pd
from pathlib import Path

# ============================================
# 1. FILE PATHS
# ============================================

base_path = Path(
    r"C:\Users\ishit\jeevan_link_ml\Data\Processed_data"
)

disaster_path = base_path / "Disaster Response Messages"

output_path = (
    base_path
    / "Assitance_Dataset_model"
    / "assistance_model_dataset_cleaned.csv"
)

split_files = {
    "train": disaster_path / "disaster_response_training_processed.csv",
    "validation": disaster_path / "disaster_response_validation_processed.csv",
    "test": disaster_path / "disaster_response_test_processed.csv"
}

label_columns = [
    "food",
    "water",
    "shelter",
    "medical_help",
    "medical_products",
    "search_and_rescue",
    "transport"
]

# ============================================
# 2. LOAD THE THREE SOURCE FILES
# ============================================

frames = []

for split_name, file_path in split_files.items():
    if not file_path.exists():
        raise FileNotFoundError(
            f"Missing source file: {file_path}"
        )

    part = pd.read_csv(file_path, low_memory=False)

    required = ["id", "message"] + label_columns
    missing = [col for col in required if col not in part.columns]

    if missing:
        raise ValueError(
            f"{file_path.name} is missing columns: {missing}"
        )

    part = part[required].copy()
    part["source_split"] = split_name
    part["source_file"] = file_path.name
    frames.append(part)

df = pd.concat(frames, ignore_index=True)

print("Rows loaded:", len(df))

# ============================================
# 3. NORMALIZE MESSAGE TEXT FOR COMPARISON
# ============================================

# Ignore case and extra whitespace when comparing messages.
df["message"] = df["message"].fillna("").astype(str).str.strip()

df["message_key"] = (
    df["message"]
    .str.casefold()
    .str.replace(r"\s+", " ", regex=True)
)

# Remove messages that are empty after cleaning.
df = df[df["message_key"] != ""].copy()

# ============================================
# 4. FIND REPEATED MESSAGES
# ============================================

duplicate_mask = df.duplicated(
    subset=["message_key"],
    keep=False
)

duplicates = df[duplicate_mask].copy()

print("Rows belonging to repeated-message groups:", len(duplicates))
print(
    "Repeated message groups:",
    duplicates["message_key"].nunique()
)

# Check whether repeated messages have conflicting labels.
label_variations = (
    duplicates.groupby("message_key")[label_columns]
    .nunique()
)

conflicting_keys = label_variations[
    (label_variations > 1).any(axis=1)
].index

conflicts = duplicates[
    duplicates["message_key"].isin(conflicting_keys)
].copy()

# Save conflicting groups separately for manual review.
conflict_path = output_path.with_name(
    "assistance_message_label_conflicts.csv"
)

conflicts.drop(columns=["message_key"]).to_csv(
    conflict_path,
    index=False
)

print("Messages with conflicting labels:", len(conflicting_keys))
print("Conflict report saved to:", conflict_path)

# ============================================
# 5. REMOVE CROSS-SPLIT DUPLICATES SAFELY
# ============================================

# If the same message has different labels, exclude the entire
# message group from this cleaned version until reviewed.
df = df[~df["message_key"].isin(conflicting_keys)].copy()

# Prefer train, then validation, then test when a message
# occurs more than once with the same labels.
split_order = {
    "train": 0,
    "validation": 1,
    "test": 2
}

df["_split_order"] = df["source_split"].map(split_order)

df = df.sort_values(
    ["_split_order"],
    kind="stable"
)

# Keep one occurrence of each normalized message.
clean_df = df.drop_duplicates(
    subset=["message_key"],
    keep="first"
).copy()

# Remove temporary comparison columns.
clean_df = clean_df.drop(
    columns=["message_key", "_split_order"]
)

# ============================================
# 6. SAVE THE CLEANED DATASET
# ============================================

output_path.parent.mkdir(parents=True, exist_ok=True)

clean_df.to_csv(output_path, index=False)

print("\nCLEANING COMPLETE")
print("-----------------")
print("Original combined rows:", sum(len(x) for x in frames))
print("Rows after cleaning:", len(clean_df))
print("Rows removed:", sum(len(x) for x in frames) - len(clean_df))
print("Remaining repeated messages:",
      clean_df["message"].str.casefold()
      .str.replace(r"\s+", " ", regex=True)
      .duplicated().sum())
print("Saved cleaned dataset to:", output_path)

print("\nFinal label counts:")
print(clean_df[label_columns].sum())

print("\nFinal dataset shape:", clean_df.shape)
```

```python
#final verdict

import pandas as pd
from pathlib import Path

# =====================================================
# 1. SET PROJECT PATHS
# =====================================================

base_path = Path(
    r"C:\Users\ishit\jeevan_link_ml\Data\Processed_data"
)

source_folder = base_path / "Disaster Response Messages"

output_folder = base_path / "Assitance_Dataset_model"

output_file = output_folder / "assistance_model_dataset_final_cleaned.csv"

conflict_file = output_folder / "assistance_message_label_conflicts.csv"

quality_report_file = output_folder / "assistance_model_dataset_final_quality_report.txt"

# Original processed dataset files
source_files = {
    "train": source_folder / "disaster_response_training_processed.csv",
    "validation": source_folder / "disaster_response_validation_processed.csv",
    "test": source_folder / "disaster_response_test_processed.csv"
}

# Assistance categories used as target labels
label_columns = [
    "food",
    "water",
    "shelter",
    "medical_help",
    "medical_products",
    "search_and_rescue",
    "transport"
]

required_columns = ["id", "message", "genre"] + label_columns

# =====================================================
# 2. CHECK SOURCE FILES
# =====================================================

for split_name, file_path in source_files.items():
    if not file_path.exists():
        raise FileNotFoundError(
            f"Missing {split_name} file: {file_path}"
        )

output_folder.mkdir(parents=True, exist_ok=True)

# =====================================================
# 3. LOAD AND SELECT DATA
# =====================================================

frames = []

for split_name, file_path in source_files.items():

    data = pd.read_csv(file_path, low_memory=False)

    missing = [
        col for col in required_columns
        if col not in data.columns
    ]

    if missing:
        raise ValueError(
            f"{file_path.name} is missing columns: {missing}"
        )

    # Keep only assistance-related features and labels
    data = data[required_columns].copy()

    # Preserve the original dataset split
    data["source_split"] = split_name
    data["source_file"] = file_path.name

    frames.append(data)

df = pd.concat(frames, ignore_index=True)

initial_rows = len(df)

print("Original combined rows:", initial_rows)

# =====================================================
# 4. CLEAN MESSAGE TEXT
# =====================================================

df["message"] = df["message"].fillna("").astype(str).str.strip()

# Remove rows without usable message text
df = df[df["message"] != ""].copy()

# Normalize text only for duplicate comparison.
# Keep the original message text in the final dataset.
df["message_key"] = (
    df["message"]
    .str.casefold()
    .str.replace(r"\s+", " ", regex=True)
)

# =====================================================
# 5. CHECK AND REMOVE INVALID LABELS
# =====================================================

# Convert labels to numeric values
for col in label_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Labels must be either 0 or 1.
# Exclude records with missing or invalid target labels.
valid_labels = (
    df[label_columns].notna().all(axis=1)
    & df[label_columns].isin([0, 1]).all(axis=1)
)

invalid_label_rows = int((~valid_labels).sum())

df = df[valid_labels].copy()

# Use integer representation for the binary labels
df[label_columns] = df[label_columns].astype(int)

# =====================================================
# 6. IDENTIFY CONFLICTING DUPLICATE MESSAGES
# =====================================================

# A conflict means identical normalized text has different
# values for at least one assistance label.

label_variations = (
    df.groupby("message_key")[label_columns]
    .nunique()
)

conflicting_keys = label_variations.index[
    (label_variations > 1).any(axis=1)
]

conflicts = df[
    df["message_key"].isin(conflicting_keys)
].copy()

# Save the conflicting records for transparency and review
conflicts.drop(columns=["message_key"]).to_csv(
    conflict_file,
    index=False
)

conflicting_group_count = len(conflicting_keys)
conflicting_row_count = len(conflicts)

# Exclude all rows from conflicting message groups.
# This avoids choosing an arbitrary label.
df = df[~df["message_key"].isin(conflicting_keys)].copy()

# =====================================================
# 7. REMOVE DUPLICATE MESSAGES
# =====================================================

# Prefer the training copy when identical messages occur
# in multiple source splits, then validation, then test.
split_order = {
    "train": 0,
    "validation": 1,
    "test": 2
}

df["_split_order"] = df["source_split"].map(split_order)

df = df.sort_values(
    "_split_order",
    kind="stable"
)

before_deduplication = len(df)

df = df.drop_duplicates(
    subset=["message_key"],
    keep="first"
).copy()

duplicate_rows_removed = before_deduplication - len(df)

# Remove temporary processing columns
df = df.drop(columns=["message_key", "_split_order"])

# =====================================================
# 8. ORGANIZE FINAL COLUMNS
# =====================================================

final_columns = [
    "id",
    "message",
    "genre",
    *label_columns,
    "source_split",
    "source_file"
]

df = df[final_columns].reset_index(drop=True)

# =====================================================
# 9. VALIDATE FINAL DATASET
# =====================================================

# Recheck the data that will actually be saved
missing_cells = int(df.isna().sum().sum())

empty_messages = int(
    df["message"].astype(str).str.strip().eq("").sum()
)

normalized_final_messages = (
    df["message"]
    .astype(str)
    .str.strip()
    .str.casefold()
    .str.replace(r"\s+", " ", regex=True)
)

duplicate_messages_remaining = int(
    normalized_final_messages.duplicated().sum()
)

duplicate_ids_remaining = int(df["id"].duplicated().sum())

invalid_labels_remaining = int(
    (~df[label_columns].isin([0, 1]).all(axis=1)).sum()
)

checks = {
    "No missing values": missing_cells == 0,
    "No empty messages": empty_messages == 0,
    "No duplicate normalized messages": duplicate_messages_remaining == 0,
    "No duplicate IDs": duplicate_ids_remaining == 0,
    "All labels are binary": invalid_labels_remaining == 0,
    "No conflicting message groups": not df["message"].astype(str)
        .str.strip().str.casefold()
        .str.replace(r"\s+", " ", regex=True)
        .isin(set(conflicts["message"].astype(str)
            .str.strip().str.casefold()
            .str.replace(r"\s+", " ", regex=True))).any(),
    "All source splits preserved": df["source_split"].notna().all()
}

# =====================================================
# 10. SAVE FINAL CSV
# =====================================================

df.to_csv(output_file, index=False)

# Reload the saved CSV to ensure it can be read correctly
saved_df = pd.read_csv(output_file, low_memory=False)

saved_shape_matches = saved_df.shape == df.shape
checks["Saved CSV matches in-memory dataset shape"] = saved_shape_matches

print("\n========== FINAL PREPROCESSING RESULTS ==========")
print("Output file:", output_file)
print("Initial rows:", initial_rows)
print("Rows with invalid labels removed:", invalid_label_rows)
print("Conflicting message groups:", conflicting_group_count)
print("Rows in conflicting groups:", conflicting_row_count)
print("Additional duplicate messages removed:", duplicate_rows_removed)
print("Final rows:", len(saved_df))
print("Final columns:", len(saved_df.columns))
print("\nRows by source split:")
print(saved_df["source_split"].value_counts())

print("\nPositive label counts:")
print(saved_df[label_columns].sum())

print("\nValidation checks:")
for check_name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'}: {check_name}")

all_passed = all(checks.values())

# =====================================================
# 11. SAVE QUALITY REPORT
# =====================================================

report_lines = [
    "Jeevan Link - Assistance Model Preprocessing Report",
    "=" * 55,
    f"Initial rows: {initial_rows}",
    f"Invalid-label rows removed: {invalid_label_rows}",
    f"Conflicting message groups: {conflicting_group_count}",
    f"Rows in conflicting groups: {conflicting_row_count}",
    f"Duplicate rows removed after conflict handling: {duplicate_rows_removed}",
    f"Final rows: {len(saved_df)}",
    f"Final columns: {len(saved_df.columns)}",
    "",
    "Validation checks:"
]

for check_name, passed in checks.items():
    report_lines.append(
        f"{'PASS' if passed else 'FAIL'}: {check_name}"
    )

report_lines += [
    "",
    "Rows by source split:",
    saved_df["source_split"].value_counts().to_string(),
    "",
    "Positive label counts:",
    saved_df[label_columns].sum().to_string(),
    "",
    "Note: Structural checks do not prove semantic correctness "
    "of the original labels or guarantee model performance."
]

quality_report_file.write_text(
    "\n".join(report_lines),
    encoding="utf-8"
)

print("\nQuality report:", quality_report_file)

if all_passed:
    print("\nSUCCESS: All automated preprocessing checks passed.")
else:
    print("\nWARNING: Some checks failed. Review the results before approval.")

print("No model was trained.")
```

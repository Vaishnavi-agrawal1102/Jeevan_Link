```python

from pathlib import Path

# Search starting point
SEARCH_DIR = Path(r"C:\Users\ishit")

# Files we need
required_files = [
    "disaster_response_training_processed.xls",
    "disaster_response_validation_processed.xls",
    "disaster_response_test_processed.xls"
]

found_files = {}

print("Searching for your dataset files...\n")

for filename in required_files:

    matches = list(SEARCH_DIR.rglob(filename))

    if matches:
        found_files[filename] = matches

        print(f"FOUND: {filename}")

        for path in matches:
            print("FULL PATH:", path)

    else:
        print(f"NOT FOUND: {filename}")

print("\n" + "=" * 60)
print("SEARCH SUMMARY")
print("=" * 60)

for filename in required_files:
    if filename in found_files:
        print(f"\n{filename}")
        print("Location:", found_files[filename][0])
    else:
        print(f"\n{filename}: NOT FOUND")
```

```python

import pandas as pd
from pathlib import Path

# ==================================================
# 1. ORIGINAL PREPROCESSED DATASET PATHS
# ==================================================

TRAIN_FILE = Path(
    r"C:\Users\ishit\Downloads\disaster_response_training_processed.xls"
)

VAL_FILE = Path(
    r"C:\Users\ishit\Downloads\disaster_response_validation_processed.xls"
)

TEST_FILE = Path(
    r"C:\Users\ishit\Downloads\disaster_response_test_processed.xls"
)

# ==================================================
# 2. LOAD DATASETS
# ==================================================

train_df = pd.read_csv(TRAIN_FILE, low_memory=False)
val_df = pd.read_csv(VAL_FILE, low_memory=False)
test_df = pd.read_csv(TEST_FILE, low_memory=False)

print("Original dataset shapes:")
print("Training:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

# ==================================================
# 3. ADD SPLIT IDENTIFIER
# ==================================================

train_df["split"] = "training"
val_df["split"] = "validation"
test_df["split"] = "test"

# ==================================================
# 4. COMBINE WITHOUT CLEANING OR CHANGING DATA
# ==================================================

assistance_dataset = pd.concat(
    [train_df, val_df, test_df],
    ignore_index=True
)

# ==================================================
# 5. SAVE ONE PREPROCESSED DATASET
# ==================================================

OUTPUT_DIR = Path(
    r"C:\Users\ishit\jeevan_link_ml\Data\Processed_data\Assitance_Dataset_model"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "assistance_dataset_preprocessed_final.csv"

assistance_dataset.to_csv(
    OUTPUT_FILE,
    index=False
)

# ==================================================
# 6. VERIFY OUTPUT
# ==================================================

print("\n========================================")
print("ASSISTANCE DATASET CREATED SUCCESSFULLY")
print("========================================")

print("Final file:", OUTPUT_FILE)
print("Total rows:", len(assistance_dataset))
print("Total columns:", len(assistance_dataset.columns))

print("\nSplit distribution:")
print(assistance_dataset["split"].value_counts())

print("\nFinal columns:")
print(assistance_dataset.columns.tolist())

print("\nSaved successfully!")
```
